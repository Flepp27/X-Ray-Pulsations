```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
XMM (EPIC) segmenting + Rayleigh + MC + wavelet + (optional) CML overlay
+ per-segment Juno JADE-E (PDS3 .LBL via pdr) spectrogram + LS + f_pe(t) output
+ per-segment Juno Waves plots (whistler + hectametric) using the same start/end.

Notes:
- XMM TIME is handled as CXC seconds with TT scale. Segment ISO strings are converted to UTC for das2.
- JADE: reads PDS3 LBL files from Data_Files/JADE Data/ and picks the one overlapping the segment.
- f_pe is saved in Hz as: [unix_time_s, f_pe_hz] to a per-segment .npy (and .csv).
- Waves whistler overlay uses that f_pe(t) for f_z and f_UHR (and optionally dynamic integration band).
"""

import os
import re
import math
import warnings
import datetime as dt

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter

from astropy.table import Table
from astropy.time import Time
from astropy.coordinates import get_body_barycentric, solar_system_ephemeris
from astropy.constants import c
import astropy.units as u

import pywt
import spiceypy as spice

import multiprocessing
from concurrent.futures import ProcessPoolExecutor

# ── Waves extras ──────────────────────────────────────────────────────────────
import pandas as pd
import matplotlib.dates as mdates
from matplotlib.colors import LogNorm
from matplotlib.ticker import MultipleLocator, AutoMinorLocator, FixedLocator
from scipy.signal import find_peaks
from matplotlib.gridspec import GridSpec
import matplotlib.patheffects as pe
from astropy.timeseries import LombScargle

# Optional das2 (Waves + ephemeris rows)
try:
    import das2
    from das2 import get_source
    HAS_DAS2 = True
except Exception:
    HAS_DAS2 = False

# Optional pdr (JADE PDS3)
try:
    import pdr
    HAS_PDR = True
except Exception:
    HAS_PDR = False

warnings.filterwarnings("ignore")

# (Optional but helpful on Apple/macOS to avoid CPU oversubscription)
os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")
os.environ.setdefault("VECLIB_MAXIMUM_THREADS", "1")
os.environ.setdefault("NUMEXPR_NUM_THREADS", "1")

# ──────────────────────────────────────────────────────────────────────────────
# CONFIG: XMM
# ──────────────────────────────────────────────────────────────────────────────
ident = ['P0801930801', 'P0801930901', 'P0801931901', 'P0882110501', 'P0882110601']
TARGET_SEGMENTS = {
    'P0801930801': 2,
    'P0801930901': 1,
    'P0801931901': 3,
    'P0882110501': 3,
    'P0882110601': 5
}

data_dir      = 'Data_Files/Fenn_Data/'
gap_threshold = 3600   # seconds

# Output locations
events_out_dir = r'/Users/fennleppard/Desktop/GitHub/X-Ray Pulsations/Segment Analysis Figures'

# Rayleigh + MC
plotit         = True
verify         = False
ntest          = 50000
startp_min     = 1.0
endp_min       = 15.0
nstep          = 1500
time_convert   = True
time_restrict  = False
tstart_min     = 104.0
tend_min       = 454.0
mc_method      = 1
lc_bin_min     = 0.25

# CML overlay on XMM lightcurve
plot_cml       = True
cml_color      = 'red'

# Multiprocessing
USE_PARALLEL_MC = True
PREFERRED_MP_START_METHOD = "fork"
TARGET_WORKERS = 10

# ──────────────────────────────────────────────────────────────────────────────
# CONFIG: WAVES (per XMM segment)
# ──────────────────────────────────────────────────────────────────────────────
MAKE_WAVES_PLOTS = True
WAVES_PRODUCTS   = ("whistler", "hect")   # ("whistler",), ("hect",), or ("whistler","hect")
WAVES_SERVER_URL = "tag:das2.org,2012:site:/uiowa/juno/wav/survey/das2"

WAVES_OUTDIR = os.path.join(events_out_dir, "Waves")

# Whistler band (Hz)
WHIS_MIN = 2e2
WHIS_MAX = 3e3

# Summed-series mode for whistler LS:
#   "fixed"   -> sum between WHIS_MIN..WHIS_MAX
#   "fz_fuhr" -> sum between f_z..f_UHR where available, else 0
WHIS_SUM_MODE = "fz_fuhr"   # or "fz_fuhr"
WHIS_FZFUHR_PAD_FRACTION = 0.0


# Whistler LS parameters
WHIS_PERIOD_MIN      = 1
WHIS_PERIOD_MAX      = 15
WHIS_PEAK_DETECT_MIN = 2.5
WHIS_NUM_PEAKS       = 4
WHIS_LS_NFREQ        = 5000

# Overlay styling
OVERLAY_LW    = 1
overlay_alpha = 0.9

# Hectametric dB spectrogram color limits
HECT_DB_VMIN = -5
HECT_DB_VMAX = 25

# Hectametric band (Hz)
HECT_MIN = 3e6
HECT_MAX = 7e6

# Hect LS parameters
HECT_PERIOD_MIN      = 1
HECT_PERIOD_MAX      = 15
HECT_PEAK_DETECT_MIN = 2.5
HECT_NUM_PEAKS       = 3
HECT_LS_NFREQ        = 5000

# ──────────────────────────────────────────────────────────────────────────────
# CONFIG: JADE (per XMM segment)
# ──────────────────────────────────────────────────────────────────────────────
MAKE_JADE_PLOTS = True
JADE_DATA_DIR   = "Data_Files/JADE Data"
JADE_OUTDIR     = os.path.join(events_out_dir, "JADE")

# Energy integration bounds (eV) for “summed series”
JADE_ENERGY_MIN_EV = 45.0
JADE_ENERGY_MAX_EV = 8000.0

# LS period window (minutes) for JADE panel
JADE_PERIOD_MIN      = 1
JADE_PERIOD_MAX      = 15
JADE_PEAK_DETECT_MIN = 2.5
JADE_NUM_PEAKS       = 4
JADE_LS_NFREQ        = 5000

# JADE-E 64 energy centers (eV)
JADE_ENERGY_BINS_EV = np.array([
    4.5731E+001, 5.3296E+001, 6.0349E+001, 6.8611E+001, 7.7419E+001,
    8.9858E+001, 1.0039E+002, 1.1494E+002, 1.2908E+002, 1.4807E+002,
    1.6538E+002, 1.8943E+002, 2.1210E+002, 2.4241E+002, 2.7057E+002,
    3.0856E+002, 3.4290E+002, 3.9006E+002, 4.3681E+002, 4.9299E+002,
    5.5223E+002, 6.2153E+002, 6.9798E+002, 7.8441E+002, 8.8190E+002,
    9.9458E+002, 1.1175E+003, 1.2632E+003, 1.4210E+003, 1.6066E+003,
    1.8068E+003, 2.0410E+003, 2.2617E+003, 2.5464E+003, 2.9011E+003,
    3.2921E+003, 3.7061E+003, 4.1751E+003, 4.7001E+003, 5.2996E+003,
    5.9765E+003, 6.7664E+003, 7.6124E+003, 8.5852E+003, 9.6487E+003,
    1.0927E+004, 1.2282E+004, 1.3872E+004, 1.5623E+004, 1.7610E+004,
    1.9783E+004, 2.2346E+004, 2.5209E+004, 2.8414E+004, 3.2039E+004,
    3.6123E+004, 4.0668E+004, 4.5886E+004, 5.1684E+004, 5.8298E+004,
    6.5677E+004, 7.4021E+004, 8.3346E+004, 9.3975E+004
], dtype=float)

JADE_N_E, JADE_N_LOOK = 64, 48
JADE_N_COLS = JADE_N_E * JADE_N_LOOK

# ──────────────────────────────────────────────────────────────────────────────
# SPICE HELPERS (CML)
# ──────────────────────────────────────────────────────────────────────────────
KERNEL_DIR = "SPICE KERNELS"
OBSERVER   = "EARTH"

def _ensure_dir(path: str):
    if path is None:
        return
    os.makedirs(path, exist_ok=True)

def _furnish_dir(dirpath: str):
    if not dirpath or not os.path.isdir(dirpath):
        return
    for fn in sorted(os.listdir(dirpath)):
        if fn.lower().endswith((".tls", ".tpc", ".tf", ".tsc",
                                ".bsp", ".bc", ".cmt", ".tm")):
            try:
                spice.furnsh(os.path.join(dirpath, fn))
            except Exception:
                pass

_furnish_dir(KERNEL_DIR)

def jupiter_cml_west_deg(et: float, observer: str = OBSERVER) -> float:
    spoint, _, _ = spice.subpnt(
        method="Near Point: Ellipsoid",
        target="JUPITER",
        et=et,
        fixref="IAU_JUPITER",
        abcorr="LT+S",
        obsrvr=observer,
    )
    _, lon_east, _ = spice.reclat(spoint)
    cml_east = (np.degrees(lon_east) % 360.0)
    return float((360.0 - cml_east) % 360.0)

# ──────────────────────────────────────────────────────────────────────────────
# XMM I/O & TIME CORRECTIONS
# ──────────────────────────────────────────────────────────────────────────────
def load_combined_times(obs_id, data_dir):
    """
    Read PN, M1, M2; correct to Jupiter–Earth light travel time; return:
      times_datetime_utc (python datetime), times_cxcsec_tt (float CXC seconds, TT)
    """
    files = [f'{data_dir}EPIC_{obs_id}_{cam}_North_RFT.fit'
             for cam in ['PN', 'M1', 'M2']]

    list_dt_utc, list_cxc = [], []

    for file in files:
        data = Table.read(file, format="fits")

        # TIME is TT-based CXC seconds
        t = Time(data['TIME'], format='cxcsec', scale='tt')

        with solar_system_ephemeris.set('de432s'):
            jup = get_body_barycentric('jupiter', t)
            xmm = get_body_barycentric('earth', t)  # approx.

        distance     = (jup - xmm).norm()
        light_time_s = (distance.to('m') / c).to('s').value
        corrected_t  = t - light_time_s * u.s

        list_dt_utc.append(corrected_t.utc.datetime)
        list_cxc.append(corrected_t.cxcsec)

    return np.hstack(list_dt_utc), np.hstack(list_cxc)

def find_gaps(times_cxcsec, threshold=3600):
    dt_arr = np.diff(times_cxcsec)
    indices = np.where(dt_arr > threshold)[0]
    gaps    = [(times_cxcsec[i], times_cxcsec[i+1]) for i in indices]
    return gaps, indices

def split_into_segments(sorted_times, gap_indices):
    split_pts = np.concatenate(([0], gap_indices + 1, [len(sorted_times)]))
    segments, seg_times_iso = [], []
    for i in range(len(split_pts) - 1):
        seg = sorted_times[split_pts[i]: split_pts[i+1]]
        if len(seg) > 1:
            segments.append(seg)
            # Convert TT-based CXC seconds -> UTC ISOT strings for external plotting/data pulls
            t0 = Time(seg[0],  format='cxcsec', scale='tt').utc.isot
            t1 = Time(seg[-1], format='cxcsec', scale='tt').utc.isot
            seg_times_iso.append((t0, t1))
    return segments, seg_times_iso

def pick_target_segment(segments, seg_times_iso, obs_id):
    selected_1based = TARGET_SEGMENTS.get(obs_id)
    if selected_1based is None:
        return segments, seg_times_iso, None
    idx = selected_1based - 1
    if 0 <= idx < len(segments):
        return [segments[idx]], [seg_times_iso[idx]], selected_1based
    print(f"Warning: requested segment {selected_1based} for {obs_id} does not exist.")
    return [], [], selected_1based

def save_unbinned_events(obs_id, seg_idx_label, segment_times_cxc):
    out_txt = os.path.join(events_out_dir, f"{obs_id}_seg_{seg_idx_label}_UNBINNED_EVENTS.txt")
    t0 = segment_times_cxc[0]
    rel = segment_times_cxc - t0
    arr = np.column_stack([segment_times_cxc, rel])
    np.savetxt(out_txt, arr, fmt="%.6f", header="ABS_CXC_SEC   REL_FROM_SEG_START_SEC")
    print(f"Saved unbinned events: {out_txt}")
    return rel  # seconds from segment start

# ──────────────────────────────────────────────────────────────────────────────
# RAYLEIGH PERIODOGRAM
# ──────────────────────────────────────────────────────────────────────────────
use_numba = True
try:
    import numba
    from packaging.version import Version
    if Version(getattr(numba, "__version__", "0.0")) > Version("0.38"):
        use_numba = True
except Exception:
    use_numba = False

if use_numba:
    @numba.jit(nopython=True)
    def _rayleigh_power_numba(omega, tf):
        nfreq = omega.size
        powv = np.zeros(nfreq, dtype=np.float64)
        for i in range(nfreq):
            ssum = 0.0
            csum = 0.0
            w = omega[i]
            for k in range(tf.size):
                ang = w * tf[k]
                ssum += math.sin(ang)
                csum += math.cos(ang)
            powv[i] = (ssum*ssum + csum*csum) / float(tf.size)
        return powv

def rayleigh_power(omega, tf):
    if use_numba:
        return _rayleigh_power_numba(omega, tf)
    s2 = np.asarray([(np.sin(w*tf).sum())**2.0 for w in omega], dtype=float)
    c2 = np.asarray([(np.cos(w*tf).sum())**2.0 for w in omega], dtype=float)
    return (s2 + c2) / float(tf.size)

# ──────────────────────────────────────────────────────────────────────────────
# MC WORKER
# ──────────────────────────────────────────────────────────────────────────────
def _mc_chunk_worker(args):
    (seed, ntrials, t0_mc, dt_window, n_events, omega, p, mc_method) = args
    rng = np.random.default_rng(seed)

    maxpow_chunk = np.empty(ntrials, dtype=float)
    pbest_chunk  = np.empty(ntrials, dtype=float)

    for ii in range(ntrials):
        if mc_method == 1:
            nfake = rng.poisson(float(n_events))
        else:
            nfake = n_events

        if nfake <= 1:
            maxpow_chunk[ii] = 0.0
            pbest_chunk[ii]  = np.nan
            continue

        tf = t0_mc + np.sort(rng.uniform(low=0.0, high=dt_window, size=nfake))
        powf = rayleigh_power(omega, tf)

        imax = np.argmax(powf)
        maxpow_chunk[ii] = powf[imax]
        pbest_chunk[ii]  = p[imax]

    return maxpow_chunk, pbest_chunk

def _get_mp_context():
    available = set(multiprocessing.get_all_start_methods())
    method = PREFERRED_MP_START_METHOD if PREFERRED_MP_START_METHOD in available else "spawn"
    return multiprocessing.get_context(method)

# ──────────────────────────────────────────────────────────────────────────────
# WAVES HELPERS
# ──────────────────────────────────────────────────────────────────────────────
def _extract_waves_arrays(d):
    f_arr = d['frequency']['center'].array
    freqs = f_arr[0, :] if f_arr.ndim > 1 else f_arr
    t_arr = d['Time']['center'].array if 'Time' in d else d['time']['center'].array
    times = t_arr[:, 0] if t_arr.ndim > 1 else t_arr
    spec  = d['spec_dens']['center'].array.squeeze()
    if spec.ndim == 2 and spec.shape[1] != freqs.size:
        spec = spec.T
    return np.array(times, dtype='datetime64[ms]'), np.asarray(freqs, float), np.asarray(spec, float)

def _compute_ls_periodogram(t_sec, y, period_min, period_max, nfreq):
    y0 = y - np.nanmean(y)
    fmin = 1.0 / (period_max * 60.0)
    fmax = 1.0 / (period_min * 60.0)
    freq = np.linspace(fmin, fmax, int(nfreq))

    ls = LombScargle(t_sec, y0, normalization="standard", center_data=False)
    power = ls.power(freq)

    periods = (1.0 / freq) / 60.0
    order = np.argsort(periods)
    periods = periods[order]
    power   = power[order]

    try:
        thr10 = ls.false_alarm_level(0.10, method="baluev",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        thr01 = ls.false_alarm_level(0.01, method="baluev",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
    except Exception:
        thr10 = ls.false_alarm_level(0.10, method="naive",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        thr01 = ls.false_alarm_level(0.01, method="naive",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
    return periods, power, thr10, thr01

def _pick_peaks(periods, power, num_peaks, min_period_label):
    peaks, _ = find_peaks(power)
    if peaks.size == 0:
        return np.array([])
    sel = peaks[np.argsort(power[peaks])][-int(num_peaks):]
    pk = periods[sel]
    pk = pk[pk >= float(min_period_label)]
    return np.sort(pk)

def _load_fpe_series_hz(fpe_path):
    """
    Expect .npy with columns [unix_time_s, f_pe_hz].
    Returns t_fpe (datetime64[ms]) and fpe_hz (float Hz).
    """
    if fpe_path is None:
        return None, None
    try:
        fpe_2d = np.load(fpe_path)
        t_unix = fpe_2d[:, 0].astype(float)
        f_hz   = fpe_2d[:, 1].astype(float)

        t_fpe = t_unix.astype("datetime64[s]").astype("datetime64[ms]")
        return t_fpe, f_hz
    except Exception:
        return None, None

def _fetch_fce_series(start_isot, end_isot):
    efc_source = das2.get_source('tag:das2.org,2012:site:/uiowa/juno/fgm/electroncyclotron/das2')
    ds_efc = efc_source.get({'time': (start_isot, end_isot)})[0]
    fce = np.asarray(ds_efc['fce']['center'].array, float)
    tt  = np.array(ds_efc['time']['center'].array, dtype='datetime64[ms]')
    ok = np.isfinite(fce) & (fce > 0)
    return tt, fce, ok

def _interp_to(t_src, y_src, t_tgt):
    ts = (t_src.astype('datetime64[ms]').astype('int64') / 1e3).astype(float)
    tt = (t_tgt.astype('datetime64[ms]').astype('int64') / 1e3).astype(float)
    ok = np.isfinite(y_src) & np.isfinite(ts)
    if np.count_nonzero(ok) < 2:
        return np.full_like(tt, np.nan, dtype=float)
    return np.interp(tt, ts[ok], np.asarray(y_src, float)[ok], left=np.nan, right=np.nan)

def plot_waves_whistler(start_isot, end_isot, title_tag="", savepath=None, fpe_series=None):
    if not HAS_DAS2:
        print("[Waves] das2 not available; skipping whistler plot.")
        return

    source = get_source(WAVES_SERVER_URL)
    whis, _, _, _ = source.get({'time': (start_isot, end_isot)})

    times, freqs, spec = _extract_waves_arrays(whis)
    if times.size < 2:
        print("[Waves] Not enough whistler data in this window.")
        return

    total_hours = (times[-1] - times[0]) / np.timedelta64(1, 'h')
    tick_interval_h = 1 if total_hours <= 12 else 2

    # Fixed whistler band
    mask_fixed = (freqs >= WHIS_MIN) & (freqs <= WHIS_MAX)

    # --------- f_pe from JADE (already computed for this segment) ----------
    t_fpe = None
    fpe_hz = None
    if fpe_series is not None:
        t_fpe, fpe_hz = fpe_series
        if t_fpe is not None:
            t_fpe = np.asarray(t_fpe, dtype="datetime64[ms]")
        if fpe_hz is not None:
            fpe_hz = np.asarray(fpe_hz, dtype=float)

    # --------- f_ce from FGM ----------
    try:
        efc_time, fce_hz, ok_fce = _fetch_fce_series(start_isot, end_isot)
    except Exception as e:
        print(f"[Waves:fce] Failed to fetch f_ce: {e}")
        efc_time, fce_hz, ok_fce = None, None, None

    # --------- Compute fz and f_UHR on f_ce time grid ----------
    fz = None
    fuhr = None

    if (t_fpe is not None) and (fpe_hz is not None) and (efc_time is not None) and (fce_hz is not None):
        # interpolate f_pe onto f_ce timestamps
        fpe_on_ce = _interp_to(t_fpe, fpe_hz, efc_time)

        good = ok_fce & np.isfinite(fpe_on_ce) & (fpe_on_ce > 0)
        if np.any(good):
            fce = np.asarray(fce_hz, float)

            # f_z = -f_ce/2 + sqrt((f_ce/2)^2 + f_pe^2)
            fz = -0.5 * fce + np.sqrt((0.5 * fce) ** 2 + fpe_on_ce ** 2)

            # f_UHR = sqrt(f_ce^2 + f_pe^2)
            fuhr = np.sqrt(fce ** 2 + fpe_on_ce ** 2)
        else:
            print("[Waves:overlay] No valid overlap to compute f_z/f_UHR (f_pe_on_ce or f_ce invalid).")
    else:
        print("[Waves:overlay] Missing f_pe series and/or f_ce series; cannot compute f_z/f_UHR.")

    # --------- Print averages over the plotted window ----------
    def _avg_in_window(t_arr, y_arr, label):
        if t_arr is None or y_arr is None:
            print(f"[Waves:avg] {label}: missing")
            return None
        t_arr = np.asarray(t_arr, dtype="datetime64[ms]")
        y_arr = np.asarray(y_arr, dtype=float)
        m = (t_arr >= times[0]) & (t_arr <= times[-1]) & np.isfinite(y_arr) & (y_arr > 0)
        if not np.any(m):
            print(f"[Waves:avg] {label}: no finite points in window")
            return None
        mean_hz = float(np.nanmean(y_arr[m]))
        print(f"[Waves:avg] {label}: mean={mean_hz:.2f} Hz  ({mean_hz/1e3:.2f} kHz)  N={int(np.sum(m))}")
        return mean_hz

    _avg_in_window(t_fpe, fpe_hz, "f_pe (JADE)")
    if efc_time is not None:
        _avg_in_window(efc_time, fce_hz, "f_ce (FGM)")
    if (efc_time is not None) and (fz is not None):
        _avg_in_window(efc_time, fz, "f_z")
    if (efc_time is not None) and (fuhr is not None):
        _avg_in_window(efc_time, fuhr, "f_UHR")

    # --------- Summed time series (fixed OR dynamic fz..fuhr) ----------
    sum_sd = np.zeros(times.size, dtype=float)

    if WHIS_SUM_MODE.lower() == "fixed":
        # Always sum fixed band
        sum_sd = np.nansum(spec[:, mask_fixed], axis=1)

    elif WHIS_SUM_MODE.lower() == "fz_fuhr":
        # Dynamic: sum between fz..fuhr where valid, else 0
        if (fz is None) or (fuhr is None) or (efc_time is None):
            print("[Waves:sum] WHIS_SUM_MODE='fz_fuhr' but fz/fuhr unavailable -> returning all zeros.")
            sum_sd[:] = 0.0
        else:
            # Interpolate fz/fuhr onto WAVES time grid
            fz_on_w = _interp_to(efc_time, fz, times)
            fu_on_w = _interp_to(efc_time, fuhr, times)

            for i in range(times.size):
                lo = fz_on_w[i]
                hi = fu_on_w[i]

                # If not valid for this time bin -> REQUIRED behavior: return 0
                if not (np.isfinite(lo) and np.isfinite(hi) and (hi > lo) and (lo > 0)):
                    sum_sd[i] = 0.0
                    continue

                # Optional padding
                if WHIS_FZFUHR_PAD_FRACTION > 0:
                    pad = WHIS_FZFUHR_PAD_FRACTION * (hi - lo)
                    lo -= pad
                    hi += pad

                # Clip to your whistler plotting band (optional but sensible)
                lo = max(lo, WHIS_MIN)
                hi = min(hi, WHIS_MAX)

                if hi <= lo:
                    sum_sd[i] = 0.0
                    continue

                mask_dyn = (freqs >= lo) & (freqs <= hi)
                sum_sd[i] = np.nansum(spec[i, mask_dyn])

    else:
        raise ValueError(f"Unknown WHIS_SUM_MODE='{WHIS_SUM_MODE}'. Use 'fixed' or 'fz_fuhr'.")


    # detrend + LS
    t0 = times[0]
    t_sec = (times - t0) / np.timedelta64(1, 's')
    dt_s  = float(np.nanmedian(np.diff(t_sec))) if t_sec.size > 2 else 60.0

    coeff = np.polyfit(t_sec, sum_sd, 1)
    resid = sum_sd - np.polyval(coeff, t_sec)

    periods_ls, power_ls, thr10, thr01 = _compute_ls_periodogram(
        t_sec, resid, WHIS_PERIOD_MIN, WHIS_PERIOD_MAX, WHIS_LS_NFREQ
    )
    mask_p = (periods_ls >= WHIS_PERIOD_MIN) & (periods_ls <= WHIS_PERIOD_MAX)
    periods_m = periods_ls[mask_p]
    pow_m     = power_ls[mask_p]
    peak_periods = _pick_peaks(periods_m, pow_m, WHIS_NUM_PEAKS, WHIS_PEAK_DETECT_MIN)

    # Plot
    fig = plt.figure(figsize=(10, 8), dpi=200)
    outer_gs = GridSpec(2, 1, figure=fig, height_ratios=[4, 1], hspace=0.5)
    inner_gs = outer_gs[0].subgridspec(2, 1, height_ratios=[3, 1], hspace=0.0)

    ax_spec = fig.add_subplot(inner_gs[0])
    ax_ts   = fig.add_subplot(inner_gs[1], sharex=ax_spec)
    ax_fft  = fig.add_subplot(outer_gs[1])

    ax_spec.pcolormesh(times, freqs, spec.T, norm=LogNorm(vmin=1e-14, vmax=1e-8),
                       cmap='turbo', shading='nearest')

    ax_spec.set_yscale('log')
    ax_spec.set_facecolor('black')
    if WHIS_SUM_MODE == 'fixed':
        
        ax_spec.plot(times, [WHIS_MIN]*len(times), '--w', lw=2, alpha=0.9)
        ax_spec.plot(times, [WHIS_MAX]*len(times), '--w', lw=2, alpha=0.9)
    else:
        pass
    ax_spec.set_ylabel('Frequency (Hz)')
    ax_spec.set_ylim(5e1, 2e4)
    ax_spec.tick_params(labelbottom=False)
    ax_spec.set_title(f'Juno Waves (Whistler) {title_tag}')

    # Overlays (only plot if at least a few points are in-range and finite)
    def _plot_overlay(ax, tt, yy, color, label):
        if tt is None or yy is None:
            print(f"[Waves:plot] {label}: missing")
            return
        tt = np.asarray(tt, dtype="datetime64[ms]")
        yy = np.asarray(yy, dtype=float)
        m = (tt >= times[0]) & (tt <= times[-1]) & np.isfinite(yy) & (yy > 0)
        if np.sum(m) < 3:
            print(f"[Waves:plot] {label}: insufficient points in window (N={int(np.sum(m))})")
            return
        # Also check y-range
        ylo, yhi = ax.get_ylim()
        in_rng = m & (yy >= ylo) & (yy <= yhi)
        if np.sum(in_rng) < 3:
            print(f"[Waves:plot] {label}: points exist but mostly outside y-limits [{ylo:.1g},{yhi:.1g}] Hz")
        ax.plot(tt[m], yy[m], lw=OVERLAY_LW, color=color, alpha=overlay_alpha,
                path_effects=[pe.Stroke(linewidth=OVERLAY_LW + 0.5, foreground='black', alpha=overlay_alpha), pe.Normal()],
                label=label)

    _plot_overlay(ax_spec, t_fpe, fpe_hz, 'magenta', r"$f_{pe}$ (JADE-E)")
    _plot_overlay(ax_spec, efc_time, fce_hz, 'dodgerblue', r"$f_{ce}$ (FGM)")
    _plot_overlay(ax_spec, efc_time, fz, 'orange', r"$f_{z}$")
    _plot_overlay(ax_spec, efc_time, fuhr, 'lime', r"$f_{UHR}$")

    ax_spec.legend(loc='upper right', frameon=True, ncols=2, fontsize='small')

    sum_sd_plot = np.where(sum_sd > 0, sum_sd, np.nan)
    ax_ts.plot(times, sum_sd_plot, linewidth=0.5, color='green')


    ax_ts.set_ylabel('Summed ESD')
    ax_ts.set_yscale('log')
    ax_ts.set_xlim(times[0], times[-1])

    major = mdates.HourLocator(interval=tick_interval_h)
    minor = mdates.MinuteLocator(interval=10)
    fmt   = mdates.DateFormatter('%H:%M')
    ax_ts.xaxis.set_major_locator(major)
    ax_ts.xaxis.set_minor_locator(minor)
    ax_ts.xaxis.set_major_formatter(fmt)

    # Ephemeris rows
    eph_src = get_source('tag:das2.org,2012:site:/uiowa/juno/ephemeris/jovicentric/das2')
    ds_eph  = eph_src.get({'time': (start_isot, end_isot, int(max(1, dt_s)))})[0]

    rad   = ds_eph['radius']['center'].array
    lon3  = ds_eph['long']['center'].array
    mlatE = ds_eph['mlat']['center'].array
    mltE  = ds_eph['mlt']['center'].array
    if mltE.dtype.kind == 'm':
        mltE = (mltE / np.timedelta64(1, 'h')).astype(float)

    t0h = times[0].astype('datetime64[h]')
    t1h = times[-1].astype('datetime64[h]') + np.timedelta64(1, 'h')
    maj = np.arange(t0h, t1h, np.timedelta64(tick_interval_h, 'h'))
    all_ticks = np.unique(np.concatenate(([times[0]], maj, [times[-1]])))
    py_ticks  = [t.tolist() for t in all_ticks]
    all_nums  = mdates.date2num(py_ticks)
    ax_ts.xaxis.set_major_locator(FixedLocator(all_nums))
    ax_ts.xaxis.set_major_formatter(fmt)

    idx_eph = [np.argmin(np.abs(times - mt)) for mt in all_ticks]

    row_lbls    = ["R$_J$", "Lon$_{III}$(°)", "MLat(°)", "MLT(h)"]
    row_arrays  = [rad, lon3, mlatE, mltE]
    row_spacing = 0.15
    for j, (label, arr) in enumerate(zip(row_lbls, row_arrays)):
        off = -row_spacing * (j + 2)
        sax = ax_ts.secondary_xaxis(off, functions=(lambda x: x, lambda x: x))
        sax.xaxis.set_major_locator(FixedLocator(all_nums))
        vals = arr[idx_eph]
        sax.set_xticklabels([f"{v:.1f}" for v in vals], rotation=0, va='top', fontsize=8)
        sax.spines['bottom'].set_visible(False)
        sax.tick_params(axis='x', length=0)
        txt = sax.set_xlabel(label, fontsize=8, labelpad=-8.5, loc='left')
        txt.set_x(-0.03)
        txt.set_ha('right')

    ax_fft.plot(periods_m, pow_m, label='LS power')
    ax_fft.axhline(thr10, ls='--', lw=1, color='k', alpha=0.8)
    ax_fft.axhline(thr01, ls='--', lw=1, color='k', alpha=0.8)
    for pk in peak_periods:
        ax_fft.axvline(pk, ls='--', color='red')
        ax_fft.text(pk, -0.05, f'{pk:.2f}', transform=ax_fft.get_xaxis_transform(),
                    rotation=90, va='top', ha='center', color='red', fontsize=7)

    ax_fft.set_xlim(WHIS_PERIOD_MIN, WHIS_PERIOD_MAX)
    ax_fft.set_xlabel('Period (minutes)')
    ax_fft.set_ylabel('Power')
    ax_fft.set_title('Whistler summed-band Lomb–Scargle')
    ax_fft.legend(loc='upper left', frameon=False, ncols=2, fontsize='small')
    ax_fft.xaxis.set_major_locator(MultipleLocator(1))
    ax_fft.xaxis.set_minor_locator(AutoMinorLocator(10))

    fig.tight_layout()
    if savepath:
        fig.savefig(savepath, bbox_inches="tight")
    plt.show()


def plot_waves_hect(start_isot, end_isot, title_tag="", savepath=None):
    if not HAS_DAS2:
        print("[Waves] das2 not available; skipping hectametric plot.")
        return

    source = get_source(WAVES_SERVER_URL)
    _, _, _, hect = source.get({'time': (start_isot, end_isot)})

    times, freqs, spec = _extract_waves_arrays(hect)
    if times.size < 2:
        print("[Waves] Not enough hectametric data in this window.")
        return

    total_hours = (times[-1] - times[0]) / np.timedelta64(1, 'h')
    tick_interval_h = 1 if total_hours <= 12 else 2

    # dB above 10th percentile background
    S_clean = spec.astype(float)
    bad = (S_clean <= 0) | (~np.isfinite(S_clean))
    S_clean[bad] = np.nan
    bg  = np.nanpercentile(S_clean, 10, axis=0)
    eps = np.nanmax(S_clean) * 1e-6
    bg  = np.where(bg <= 0, eps, bg)
    dB_hect = 20 * np.log10(spec / bg[None, :])

    band_mask = (freqs >= HECT_MIN) & (freqs <= HECT_MAX)
    dB_band   = np.nanmean(dB_hect[:, band_mask], axis=1)

    t0 = times[0]
    t_sec = (times - t0) / np.timedelta64(1, 's')
    dt_s  = float(np.nanmedian(np.diff(t_sec))) if t_sec.size > 2 else 60.0

    coeff = np.polyfit(t_sec, dB_band, 1)
    resid = dB_band - np.polyval(coeff, t_sec)

    periods_ls, power_ls, thr10, thr01 = _compute_ls_periodogram(
        t_sec, resid, HECT_PERIOD_MIN, HECT_PERIOD_MAX, HECT_LS_NFREQ
    )
    mask_p = (periods_ls >= HECT_PERIOD_MIN) & (periods_ls <= HECT_PERIOD_MAX)
    periods_m = periods_ls[mask_p]
    pow_m     = power_ls[mask_p]
    peak_periods = _pick_peaks(periods_m, pow_m, HECT_NUM_PEAKS, HECT_PEAK_DETECT_MIN)

    fig = plt.figure(figsize=(10, 8), dpi=200)
    outer_gs = GridSpec(2, 1, figure=fig, height_ratios=[3, 1], hspace=0.5)

    inner_gs = outer_gs[0].subgridspec(2, 1, height_ratios=[3, 1], hspace=0.0)
    ax_spec = fig.add_subplot(inner_gs[0])
    ax_ts   = fig.add_subplot(inner_gs[1], sharex=ax_spec)
    ax_fft  = fig.add_subplot(outer_gs[1])

    ax_spec.pcolormesh(times, freqs, dB_hect.T, cmap='turbo', shading='nearest',
                       vmin=HECT_DB_VMIN, vmax=HECT_DB_VMAX)
    ax_spec.set_yscale('log')
    ax_spec.set_facecolor('black')
    ax_spec.plot(times, [HECT_MIN]*len(times), '--w', lw=2)
    ax_spec.plot(times, [HECT_MAX]*len(times), '--w', lw=2)
    ax_spec.set_ylabel('Frequency (Hz)')
    ax_spec.set_title(f'Hectametric dB Spectrogram {title_tag}')
    ax_spec.tick_params(labelbottom=False)

    ax_ts.plot(times, dB_band, linewidth=0.8)
    ax_ts.set_ylabel(f'Mean dB [{HECT_MIN/1e6:.0f}-{HECT_MAX/1e6:.0f} MHz]')
    ax_ts.set_xlim(times[0], times[-1])

    # Unified ticks
    t0h = times[0].astype('datetime64[h]')
    t1h = times[-1].astype('datetime64[h]') + np.timedelta64(1, 'h')
    maj = np.arange(t0h, t1h, np.timedelta64(tick_interval_h, 'h'))
    all_ticks = np.unique(np.concatenate(([times[0]], maj, [times[-1]])))
    py_ticks  = [t.tolist() for t in all_ticks]
    all_nums  = mdates.date2num(py_ticks)
    fmt = mdates.DateFormatter('%H:%M')
    ax_ts.xaxis.set_major_locator(FixedLocator(all_nums))
    ax_ts.xaxis.set_major_formatter(fmt)
    ax_ts.xaxis.set_minor_locator(mdates.MinuteLocator(interval=10))

    # Ephemeris rows
    eph_src = get_source('tag:das2.org,2012:site:/uiowa/juno/ephemeris/jovicentric/das2')
    ds_eph  = eph_src.get({'time': (start_isot, end_isot, int(max(1, dt_s)))})[0]

    rad   = ds_eph['radius']['center'].array
    lon3  = ds_eph['long']['center'].array
    mlatE = ds_eph['mlat']['center'].array
    mltE  = ds_eph['mlt']['center'].array
    if mltE.dtype.kind == 'm':
        mltE = (mltE / np.timedelta64(1, 'h')).astype(float)

    row_lbls    = ["R$_J$", "Lon$_{III}$(°)", "MLat(°)", "MLT(h)"]
    row_arrays  = [rad, lon3, mlatE, mltE]
    row_spacing = 0.15
    for j, (label, arr) in enumerate(zip(row_lbls, row_arrays)):
        off = -row_spacing * (j + 2)
        sax = ax_ts.secondary_xaxis(off, functions=(lambda x: x, lambda x: x))
        sax.xaxis.set_major_locator(FixedLocator(all_nums))
        vals = arr[[np.argmin(np.abs(times - t64)) for t64 in all_ticks]]
        sax.set_xticklabels([f"{v:.1f}" for v in vals], rotation=0, va='top', fontsize=8)
        sax.spines['bottom'].set_visible(False)
        sax.tick_params(axis='x', length=0)
        txt = sax.set_xlabel(label, fontsize=8, labelpad=-8.5, loc='left')
        txt.set_x(-0.025)
        txt.set_ha('right')

    # LS panel
    ax_fft.plot(periods_m, pow_m, linewidth=1, label='LS power')
    ax_fft.axhline(thr10, ls='--', lw=1, color='k', alpha=0.8)
    ax_fft.axhline(thr01, ls='--', lw=1, color='k', alpha=0.8)
    for pk in peak_periods:
        ax_fft.axvline(pk, ls='--', color='red')
        ax_fft.text(pk, -0.05, f'{pk:.2f} min', transform=ax_fft.get_xaxis_transform(),
                    rotation=90, va='top', ha='center', color='red', fontsize=7)

    ax_fft.set_xlim(HECT_PERIOD_MIN, HECT_PERIOD_MAX)
    ax_fft.set_xlabel('Period (minutes)')
    ax_fft.set_ylabel('Power')
    ax_fft.set_title('Hect band-avg Lomb–Scargle')
    ax_fft.legend(loc='upper left', frameon=False)
    ax_fft.xaxis.set_major_locator(MultipleLocator(1))
    ax_fft.xaxis.set_minor_locator(AutoMinorLocator(10))

    fig.tight_layout()
    if savepath:
        fig.savefig(savepath, bbox_inches="tight")
    plt.show()

def plot_waves_for_interval(start_isot, end_isot, tag, fpe_path_hz=None):
    if not MAKE_WAVES_PLOTS:
        return
    if not HAS_DAS2:
        print("[Waves] MAKE_WAVES_PLOTS=True but das2 unavailable; skipping Waves.")
        return

    _ensure_dir(WAVES_OUTDIR)

    if "whistler" in WAVES_PRODUCTS:
        out = None if WAVES_OUTDIR is None else os.path.join(WAVES_OUTDIR, f"WAVES_WHIS_{tag}.png")
        plot_waves_whistler(start_isot, end_isot, title_tag=tag, savepath=out, fpe_path_hz=fpe_path_hz)

    if "hect" in WAVES_PRODUCTS:
        out = None if WAVES_OUTDIR is None else os.path.join(WAVES_OUTDIR, f"WAVES_HECT_{tag}.png")
        plot_waves_hect(start_isot, end_isot, title_tag=tag, savepath=out)

# ──────────────────────────────────────────────────────────────────────────────
# JADE HELPERS (PDS3 LBL via pdr)
# ──────────────────────────────────────────────────────────────────────────────
def jade_energy_edges_from_centers(e):
    """Geometric midpoints; extrapolated outer edges."""
    e = np.asarray(e, float)
    mids = np.sqrt(e[:-1] * e[1:])
    return np.concatenate([[e[0]**2 / mids[0]], mids, [e[-1]**2 / mids[-1]]])

# Cache per LBL: time ranges (fast selection) and optional loaded arrays
_JADE_FILE_TIMERANGE_CACHE = {}  # path -> (tmin, tmax) as pandas Timestamp
_JADE_LOADED_CACHE = {}          # path -> (t_edges, t_centers, e_sorted, e_edges, Zp)

def _jade_list_lbls(data_dir):
    if not os.path.isdir(data_dir):
        return []
    lbls = []
    for root, _, files in os.walk(data_dir):
        for fn in files:
            if fn.lower().endswith(".lbl"):
                lbls.append(os.path.join(root, fn))
    return sorted(lbls)

def _jade_get_file_timerange(lbl_path):
    if lbl_path in _JADE_FILE_TIMERANGE_CACHE:
        return _JADE_FILE_TIMERANGE_CACHE[lbl_path]

    ds = pdr.read(lbl_path)
    df = ds["TABLE"]

    t_lo = pd.to_datetime(df["DIM0_UTC_LOWER"].astype(str), format="%Y-%jT%H:%M:%S.%f", utc=False, errors="coerce")
    t_hi = pd.to_datetime(df["DIM0_UTC_UPPER"].astype(str), format="%Y-%jT%H:%M:%S.%f", utc=False, errors="coerce")

    t_lo = t_lo.dropna()
    t_hi = t_hi.dropna()
    if len(t_lo) == 0 or len(t_hi) == 0:
        _JADE_FILE_TIMERANGE_CACHE[lbl_path] = (None, None)
        return None, None

    tmin = t_lo.min()
    tmax = t_hi.max()
    _JADE_FILE_TIMERANGE_CACHE[lbl_path] = (tmin, tmax)
    return tmin, tmax

def jade_find_lbl_for_interval(start_isot, end_isot, data_dir=JADE_DATA_DIR):
    """Pick the JADE LBL file that overlaps [start,end]. Returns path or None."""
    if not HAS_PDR:
        return None

    t0 = pd.to_datetime(start_isot)
    t1 = pd.to_datetime(end_isot)
    if t1 < t0:
        t0, t1 = t1, t0

    candidates = []
    for p in _jade_list_lbls(data_dir):
        tmin, tmax = _jade_get_file_timerange(p)
        if tmin is None or tmax is None:
            continue
        if (tmax >= t0) and (tmin <= t1):
            # overlap exists
            overlap = (min(tmax, t1) - max(tmin, t0)).total_seconds()
            candidates.append((overlap, p))

    if not candidates:
        return None
    candidates.sort(key=lambda x: x[0], reverse=True)
    return candidates[0][1]

def jade_load_cube(lbl_path):
    """
    Load one JADE LBL into:
      t_edges (DatetimeIndex length nt+1),
      t_centers (DatetimeIndex length nt),
      e_sorted (64,),
      e_edges (65,),
      Zp (nt,64) = mean across look directions
    """
    if lbl_path in _JADE_LOADED_CACHE:
        return _JADE_LOADED_CACHE[lbl_path]

    ds = pdr.read(lbl_path)
    df = ds["TABLE"]

    t_lo = pd.to_datetime(df["DIM0_UTC_LOWER"].astype(str), format="%Y-%jT%H:%M:%S.%f", utc=False, errors="coerce")
    t_hi = pd.to_datetime(df["DIM0_UTC_UPPER"].astype(str), format="%Y-%jT%H:%M:%S.%f", utc=False, errors="coerce")

    pat = re.compile(r"^DATA_(\d+)$", re.IGNORECASE)
    pairs = []
    for c in df.columns:
        m = pat.match(str(c))
        if m:
            pairs.append((int(m.group(1)), c))
    pairs.sort(key=lambda x: x[0])
    data_cols = [c for _, c in pairs]
    if len(data_cols) != JADE_N_COLS:
        raise ValueError(f"[JADE] Expected {JADE_N_COLS} DATA columns, got {len(data_cols)} in {os.path.basename(lbl_path)}")

    raw = df[data_cols].to_numpy(dtype=float)
    raw[raw <= -9.0e5] = np.nan

    nt = len(df)
    cube = raw.reshape(nt, JADE_N_E, JADE_N_LOOK)      # (T,64,48)
    Z_mean = np.nanmean(cube, axis=2)                  # (T,64)

    mask = t_lo.notna() & t_hi.notna()
    t_lo = t_lo[mask].reset_index(drop=True)
    t_hi = t_hi[mask].reset_index(drop=True)
    Z_mean = Z_mean[mask.values, :]

    ord_idx = np.argsort(t_lo.values)
    t_lo = t_lo.iloc[ord_idx].reset_index(drop=True)
    t_hi = t_hi.iloc[ord_idx].reset_index(drop=True)
    Z_mean = Z_mean[ord_idx, :]

    t_edges = pd.DatetimeIndex(np.concatenate([t_lo.values[:1], t_hi.values]))
    t_centers = t_lo + (t_hi - t_lo) / 2

    sorter = np.argsort(JADE_ENERGY_BINS_EV)
    e_sorted = JADE_ENERGY_BINS_EV[sorter]
    Zp = Z_mean[:, sorter]
    e_edges = jade_energy_edges_from_centers(e_sorted)

    _JADE_LOADED_CACHE[lbl_path] = (t_edges, t_centers, e_sorted, e_edges, Zp)
    return t_edges, t_centers, e_sorted, e_edges, Zp

def _jade_interpolate_nans_time(times_ns, y):
    y = np.asarray(y, float)
    if np.all(np.isfinite(y)):
        return y
    s = pd.Series(y, index=pd.to_datetime(times_ns))
    s = s.interpolate(method="time").bfill().ffill()
    return s.to_numpy(dtype=float)

def _jade_summarize_between_lines(e_sorted, Zp_win, emin_ev, emax_ev):
    mask_e = (e_sorted >= emin_ev) & (e_sorted <= emax_ev)
    if not np.any(mask_e):
        raise ValueError("[JADE] Energy mask empty; check JADE_ENERGY_MIN_EV/JADE_ENERGY_MAX_EV")
    summed = np.nansum(Zp_win[:, mask_e], axis=1)
    return summed, mask_e

def jade_make_segment_analysis_and_fpe(
    start_isot,
    end_isot,
    tag,
    lbl_path=None,
    emin_ev=JADE_ENERGY_MIN_EV,
    emax_ev=JADE_ENERGY_MAX_EV,
    period_min=JADE_PERIOD_MIN,
    period_max=JADE_PERIOD_MAX,
    peak_detect_min=JADE_PEAK_DETECT_MIN,
    num_peaks=JADE_NUM_PEAKS,
    save_png=True
):
    """
    Runs JADE spectrogram + summed series + LS, and computes n_e(t) and f_pe(t).
    Saves:
      - JADE segment analysis PNG
      - f_pe series .npy/.csv (Hz): columns [unix_time_s, f_pe_hz]
    Returns:
      fpe_path_npy or None if failed.
    """
    if not MAKE_JADE_PLOTS:
        return None
    if not HAS_PDR:
        print("[JADE] MAKE_JADE_PLOTS=True but pdr is unavailable; skipping JADE.")
        return None

    _ensure_dir(JADE_OUTDIR)

    if lbl_path is None:
        lbl_path = jade_find_lbl_for_interval(start_isot, end_isot, JADE_DATA_DIR)

    if lbl_path is None:
        print(f"[JADE] No LBL found in '{JADE_DATA_DIR}' overlapping {start_isot} to {end_isot}")
        return None

    try:
        t_edges, t_centers, e_sorted, e_edges, Zp = jade_load_cube(lbl_path)
    except Exception as e:
        print(f"[JADE] Failed loading {os.path.basename(lbl_path)}: {e}")
        return None

    t0 = pd.to_datetime(start_isot)
    t1 = pd.to_datetime(end_isot)
    if t1 < t0:
        t0, t1 = t1, t0

    mask_t = (t_centers >= t0) & (t_centers <= t1)
    if not np.any(mask_t):
        print(f"[JADE] Window selects no samples within {os.path.basename(lbl_path)}")
        return None

    idxs = np.where(mask_t.values)[0]
    idx0 = int(idxs[0])
    idx1 = int(idxs[-1])

    times_win = t_centers[mask_t].to_numpy(dtype="datetime64[ns]")  # (Nt,)
    Z_win     = Zp[mask_t.values, :]                                 # (Nt,64)

    # Summed series
    avg_counts, mask_e = _jade_summarize_between_lines(e_sorted, Z_win, emin_ev, emax_ev)
    avg_counts = _jade_interpolate_nans_time(times_win, avg_counts)

    # Detrend + residual
    t0c = times_win[0]
    t_sec = (times_win - t0c) / np.timedelta64(1, "s")
    if len(t_sec) < 4:
        print("[JADE] Too few samples for LS in this window.")
        return None

    dt_s = float(np.median(np.diff(t_sec)))
    if not np.isfinite(dt_s) or dt_s <= 0:
        print("[JADE] Invalid cadence for this window.")
        return None

    coeff = np.polyfit(t_sec, avg_counts, 1)
    trend = np.polyval(coeff, t_sec)
    residual = avg_counts - trend
    residual = residual - np.mean(residual)

    # Lomb–Scargle
    fmin = 1.0 / (period_max * 60.0)
    fmax = 1.0 / (period_min * 60.0)
    freq = np.linspace(fmin, fmax, JADE_LS_NFREQ)

    ls = LombScargle(t_sec, residual, normalization="standard", center_data=False)
    power = ls.power(freq)

    periods_raw = (1.0 / freq) / 60.0
    order = np.argsort(periods_raw)
    periods_ls = periods_raw[order]
    power_ls   = power[order]

    # Peak picking
    peaks, _ = find_peaks(power_ls)
    if peaks.size:
        sel = peaks[np.argsort(power_ls[peaks])][-num_peaks:]
        sel = np.sort(sel)
        peak_periods = periods_ls[sel]
        peak_periods = peak_periods[peak_periods >= peak_detect_min]
    else:
        peak_periods = np.array([])

    # FAP thresholds
    try:
        thr_10 = ls.false_alarm_level(0.10, method="baluev",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        thr_01 = ls.false_alarm_level(0.01, method="baluev",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        fap_method = "baluev"
    except Exception:
        thr_10 = ls.false_alarm_level(0.10, method="naive",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        thr_01 = ls.false_alarm_level(0.01, method="naive",
                                     minimum_frequency=fmin, maximum_frequency=fmax)
        fap_method = "naive"

    # Plot
    fig = plt.figure(figsize=(10.5, 9), dpi=250)
    outer_gs = GridSpec(2, 1, figure=fig, height_ratios=[3, 1], hspace=0.55)
    inner_gs = outer_gs[0].subgridspec(2, 1, height_ratios=[2, 1], hspace=0.0)

    ax_spec = fig.add_subplot(inner_gs[0])
    ax_avg  = fig.add_subplot(inner_gs[1], sharex=ax_spec)
    ax_ls   = fig.add_subplot(outer_gs[1])

    # Use windowed edges
    t_edges_win = t_edges[idx0: idx1 + 2]

    Zpos = Z_win[np.isfinite(Z_win) & (Z_win > 0)]
    if Zpos.size == 0:
        print("[JADE] No positive values to plot in this window.")
        plt.close(fig)
        return None

    vmin, vmax = np.nanpercentile(Zpos, [5, 95])
    vmin = max(vmin, np.nextafter(0.0, 1.0))

    ax_spec.pcolormesh(
        t_edges_win, e_edges, Z_win.T, shading="auto",
        norm=LogNorm(vmin=vmin, vmax=vmax), cmap="turbo"
    )
    ax_spec.set_yscale("log")
    ax_spec.set_facecolor("black")
    if WHIS_SUM_MODE == 'fixed':
        ax_spec.plot([times_win[0], times_win[-1]], [emin_ev, emin_ev], "--w", lw=2)
        ax_spec.plot([times_win[0], times_win[-1]], [emax_ev, emax_ev], "--w", lw=2)
    else:
        pass
    
    ax_spec.set_ylabel("Energy (eV)")
    ax_spec.set_title(f"JADE-E LRS spectrogram + segment analysis\n{tag}")
    ax_spec.tick_params(labelbottom=False)

    ax_avg.plot(times_win, avg_counts, label="Original", color="C0", lw=1.0)
    ax_avg.plot(times_win, trend, linestyle="--", color="C1", label="Trend", lw=1.0)
    ax_avg.plot(times_win, residual, color="C2", label="Detrended", lw=1.0)
    ax_avg.set_ylabel(f"Summed {emin_ev:.0f}–{emax_ev:.0f} eV")
    ax_avg.legend(frameon=False, fontsize=9)

    # Ticks
    total_hours = (times_win[-1] - times_win[0]) / np.timedelta64(1, "h")
    tick_interval_h = 1 if total_hours <= 12 else 2
    eph_step = np.timedelta64(30, "m") if total_hours <= 4 else np.timedelta64(int(tick_interval_h), "h")

    t0h = times_win[0].astype("datetime64[h]")
    t1h = times_win[-1].astype("datetime64[h]") + np.timedelta64(1, "h")
    maj = np.arange(t0h, t1h, eph_step)
    all_ticks = np.unique(np.concatenate(([times_win[0]], maj, [times_win[-1]])))
    py_ticks  = [pd.to_datetime(t).to_pydatetime() for t in all_ticks]
    tick_nums = mdates.date2num(py_ticks)

    ax_avg.xaxis.set_major_locator(FixedLocator(tick_nums))
    ax_avg.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
    ax_avg.xaxis.set_minor_locator(mdates.MinuteLocator(interval=10))

    # Ephemeris rows (optional, needs das2)
    if HAS_DAS2:
        try:
            eph_src = get_source("tag:das2.org,2012:site:/uiowa/juno/ephemeris/jovicentric/das2")
            ds_eph  = eph_src.get({"time": (start_isot, end_isot, int(max(1, round(dt_s))))})[0]

            eph_t  = ds_eph["time"]["center"].array
            rad    = ds_eph["radius"]["center"].array
            lon3   = ds_eph["long"]["center"].array
            mlatE  = ds_eph["mlat"]["center"].array
            mltE   = ds_eph["mlt"]["center"].array
            ioE    = ds_eph["io_phase"]["center"].array
            if getattr(mltE, "dtype", None) is not None and mltE.dtype.kind == "m":
                mltE = (mltE / np.timedelta64(1, "h")).astype(float)

            row_lbls   = ["R$_J$", "Lon$_{III}$(°)", "MLat(°)", "MLT(h)", "Io phase(°)"]
            row_arrays = [rad, lon3, mlatE, mltE, ioE]
            row_spacing = 0.12

            idxs_eph = [int(np.argmin(np.abs(eph_t - np.datetime64(t)))) for t in all_ticks]
            for j, (label, arr) in enumerate(zip(row_lbls, row_arrays)):
                off = -row_spacing * (j + 2)
                sax = ax_avg.secondary_xaxis(off, functions=(lambda x: x, lambda x: x))
                sax.xaxis.set_major_locator(FixedLocator(tick_nums))
                vals = arr[idxs_eph]
                sax.set_xticklabels([f"{float(v):.1f}" for v in vals], rotation=0, va="top", fontsize=8)
                sax.spines["bottom"].set_visible(False)
                sax.tick_params(axis="x", length=0)
                txt = sax.set_xlabel(label, fontsize=8, labelpad=-8.5, loc="left")
                txt.set_x(-0.025)
                txt.set_ha("right")
        except Exception as e:
            print(f"[JADE] Ephemeris rows failed: {e}")

    # LS panel
    ax_ls.plot(periods_ls, power_ls, color="C0", lw=1, label="LS power")
    ax_ls.axhline(thr_10, ls="--", lw=1, color="k", alpha=0.8, label=f"FAP 10% ({fap_method})")
    ax_ls.axhline(thr_01, ls="--", lw=1, color="k", alpha=0.8, label=f"FAP 1% ({fap_method})")
    for ppk in peak_periods:
        ax_ls.axvline(ppk, ls="--", color="red", lw=1)
        ax_ls.text(ppk, -0.05, f"{ppk:.2f}", transform=ax_ls.get_xaxis_transform(),
                   rotation=90, va="top", ha="center", color="red", fontsize=7)

    ax_ls.set_xlim(period_min, period_max)
    ax_ls.set_xlabel("Period (minutes)")
    ax_ls.set_ylabel("Power")
    ax_ls.set_title("JADE summed-series Lomb–Scargle")
    ax_ls.xaxis.set_major_locator(MultipleLocator(1))
    ax_ls.xaxis.set_minor_locator(AutoMinorLocator(10))
    ax_ls.legend(loc="upper right", frameon=True, fontsize=7)

    fig.tight_layout()

    if save_png:
        out_png = os.path.join(JADE_OUTDIR, f"JADE_SEG_{tag}.png")
        fig.savefig(out_png, dpi=250, bbox_inches="tight")
        print(f"[JADE] Saved → {out_png}")

    plt.show()

    # ---------------- n_e(t) and f_pe(t) ----------------
    E_ev  = e_sorted.astype(float)
    dE_ev = np.diff(e_edges).astype(float)  # (64,)

    j_per_eV = Z_win / dE_ev[np.newaxis, :]

    me = 9.10938356e-31
    qe = 1.602176634e-19
    v_cms = np.sqrt(2.0 * (E_ev * qe) / me) * 100.0  # cm/s

    emask = (E_ev >= 50.0) & (E_ev <= 20000.0)

    integrand = j_per_eV[:, emask] / v_cms[np.newaxis, emask]
    ne_cm3 = 4.0 * np.pi * np.trapz(integrand, x=E_ev[emask], axis=1)

    # f_pe in Hz
    fpe_hz = 8980.0 * np.sqrt(np.clip(ne_cm3, 0, None))

    # IMPORTANT: fill any gaps so we can interpolate onto f_ce cleanly
    # (uses the same time-interp helper already in your JADE code)
    fpe_hz = _jade_interpolate_nans_time(times_win, fpe_hz)

    # return in datetime64[ms] for easy masking with Waves arrays
    t_fpe_ms = times_win.astype("datetime64[ms]")

    # quick stats for this segment
    ok = np.isfinite(fpe_hz) & (fpe_hz > 0)
    if np.any(ok):
        print(f"[JADE:fpe] {tag}  mean={np.nanmean(fpe_hz[ok])/1e3:.2f} kHz  "
              f"med={np.nanmedian(fpe_hz[ok])/1e3:.2f} kHz  "
              f"min={np.nanmin(fpe_hz[ok])/1e3:.2f} kHz  max={np.nanmax(fpe_hz[ok])/1e3:.2f} kHz  "
              f"N={ok.sum()}")
    else:
        print(f"[JADE:fpe] {tag}  no finite positive f_pe samples")

    return t_fpe_ms, fpe_hz

# ──────────────────────────────────────────────────────────────────────────────
# MAIN ANALYSIS: RAYLEIGH + WAVELET + PLOTTING + MC
# ──────────────────────────────────────────────────────────────────────────────
def run_rayleigh_and_mc(t_rel_seconds,
                        abs_t0=None,
                        startp_min=2.0, endp_min=100.0, nstep=1500,
                        time_convert=True, time_restrict=False,
                        tstart_min=0.0, tend_min=0.0,
                        ntest=10000, mc_method=2,
                        verify=False, plotit=True,
                        fig_tag="",
                        lc_bin_min=1.0,
                        n_scales_wavelet=80,
                        plot_cml_overlay=False):

    t = np.array(t_rel_seconds, dtype=float)
    if time_convert:
        t = t / 60.0

    if time_restrict:
        mask = (t >= tstart_min) & (t <= tend_min)
        t = t[mask]
        if t.size < 2:
            raise ValueError("After restriction, fewer than 2 events remain.")

    # lightcurve for wavelet
    if t.size > 0:
        t0_lc, t1_lc = t.min(), t.max()
        if t1_lc == t0_lc:
            t1_lc = t0_lc + lc_bin_min

        nbins = max(1, int(np.ceil((t1_lc - t0_lc) / lc_bin_min)))
        lc_edges   = np.linspace(t0_lc, t1_lc, nbins + 1)
        lc_counts, _ = np.histogram(t, bins=lc_edges)
        lc_centers = 0.5 * (lc_edges[:-1] + lc_edges[1:])
    else:
        lc_centers = np.array([])
        lc_counts  = np.array([])

    if abs_t0 is not None and lc_centers.size > 0:
        lc_datetimes = np.array([abs_t0 + dt.timedelta(minutes=float(m)) for m in lc_centers])
    else:
        lc_datetimes = None

    fmin = 1.0 / endp_min
    fmax = 1.0 / startp_min
    f = np.linspace(fmin, fmax, int(nstep))
    p = 1.0 / f
    omega = 2.0 * np.pi * f

    pow_data = rayleigh_power(omega, t)
    ibest   = np.argmax(pow_data)
    fbest   = f[ibest]
    pbest   = p[ibest]
    powmax  = pow_data[ibest]

    print(f"[Rayleigh] Best period = {pbest:.4f} min | best freq = {fbest:.6f} 1/min | Max power = {powmax:.6f}")

    # wavelet
    wavelet_periods = None
    wavelet_power   = None
    if lc_centers.size > 3:
        wavename = 'morl'
        dt_min = lc_bin_min
        wavelet_periods = np.linspace(startp_min, endp_min, n_scales_wavelet)

        fc = pywt.central_frequency(wavename)
        scales = fc * wavelet_periods / dt_min

        y = lc_counts.astype(float) - np.mean(lc_counts)
        coeffs, _ = pywt.cwt(y, scales, wavename, sampling_period=dt_min)
        wavelet_power = np.abs(coeffs) ** 2
    else:
        print("[Wavelet] Not enough lightcurve bins for wavelet analysis.")

    # plotting
    if plotit:
        nrows = 4 if wavelet_power is not None else 3
        fig, axes = plt.subplots(nrows, 1, figsize=(8, 3*nrows + 2), dpi=150)

        ax_lc    = axes[0]
        ax_per   = axes[1]
        ax_freq  = axes[2]
        ax_wvlt  = axes[3] if nrows == 4 else None

        if lc_counts.size > 0:
            if lc_datetimes is not None:
                ax_lc.step(lc_datetimes, lc_counts, where='mid')
                ax_lc.set_xlabel('UTC Time')
                ax_lc.xaxis.set_major_formatter(DateFormatter("%H:%M"))
                ax_lc.tick_params(axis='x', rotation=45)
            else:
                ax_lc.step(lc_centers, lc_counts, where='mid')
                ax_lc.set_xlabel('Time from seg start (min)')

            ax_lc.set_ylabel('Counts / bin')
            ax_lc.set_title(f'Lightcurve {fig_tag}')

            if plot_cml_overlay and (lc_datetimes is not None):
                try:
                    lc_et = np.array([spice.datetime2et(t_) for t_ in lc_datetimes])
                    lc_cml = np.array([jupiter_cml_west_deg(et) for et in lc_et])

                    ax_cml = ax_lc.twinx()
                    ax_cml.plot(lc_datetimes, lc_cml, color=cml_color, lw=1.6,
                                alpha=0.5, linestyle='--')
                    ax_cml.set_ylabel("CML (°)", color=cml_color)
                    ax_cml.tick_params(axis='y', labelcolor=cml_color)
                    ax_cml.grid(False)
                except Exception as e:
                    print(f"[CML] Failed to compute CML overlay: {e}")
        else:
            ax_lc.text(0.5, 0.5, 'No events', ha='center', va='center', transform=ax_lc.transAxes)
            ax_lc.set_axis_off()

        ax_per.plot(p, pow_data, '-k', lw=1)
        ax_per.scatter(p, pow_data, s=12)
        ax_per.set_xlabel('Period (min)')
        ax_per.set_ylabel('Rayleigh Power')
        ax_per.set_title(f'Rayleigh Periodogram {fig_tag}')
        ax_per.axvline(pbest, ls='--', lw=1, color='gray')
        ax_per.annotate(f'{pbest:.2f} min', xy=(pbest, powmax),
                        xytext=(pbest, powmax * 1.05), ha='center', fontsize=8)
        ax_per.set_ylim(0, powmax * 1.2)

        ax_freq.plot(f, pow_data, '-k', lw=1)
        ax_freq.scatter(f, pow_data, s=12)
        ax_freq.set_xlabel('Frequency (1/min)')
        ax_freq.set_ylabel('Rayleigh Power')
        ax_freq.set_title(f'Rayleigh Periodogram (frequency) {fig_tag}')
        ax_freq.axvline(fbest, ls='--', lw=1, color='gray')
        ax_freq.annotate(f'{fbest:.4f} 1/min', xy=(fbest, powmax),
                         xytext=(fbest, powmax * 1.05), ha='center', fontsize=8)
        ax_freq.set_ylim(0, powmax * 1.2)

        if ax_wvlt is not None and wavelet_power is not None:
            if lc_datetimes is not None:
                T, Pgrid = np.meshgrid(lc_datetimes, wavelet_periods)
            else:
                T, Pgrid = np.meshgrid(lc_centers, wavelet_periods)

            ax_wvlt.pcolormesh(T, Pgrid, wavelet_power, shading='auto', cmap='seismic')
            ax_wvlt.set_ylabel('Period (min)')
            if lc_datetimes is not None:
                ax_wvlt.set_xlabel('UTC Time')
                ax_wvlt.xaxis.set_major_formatter(DateFormatter("%H:%M"))
                ax_wvlt.tick_params(axis='x', rotation=45)
            else:
                ax_wvlt.set_xlabel('Time from seg start (min)')
            ax_wvlt.set_title(f'Wavelet Power (Morlet) {fig_tag}')
            ax_wvlt.axhline(pbest, color='white', ls='--', lw=1)
            ax_wvlt.invert_yaxis()

        fig.tight_layout()
        plt.show()

    # Monte Carlo
    if time_restrict:
        t0_mc, t1_mc = tstart_min, tend_min
    else:
        t0_mc, t1_mc = t.min(), t.max()

    dt_window = t1_mc - t0_mc
    if dt_window <= 0:
        raise ValueError("Non-positive time window for Monte Carlo.")
    n_events = t.size
    if n_events < 2:
        raise ValueError("Need at least 2 events for Monte Carlo significance.")

    maxpow_fake = np.empty(ntest, dtype=float)
    pbest_fake  = np.empty(ntest, dtype=float)

    n_cpu = os.cpu_count() or 1
    target_workers = TARGET_WORKERS or n_cpu
    n_workers = min(target_workers, n_cpu, ntest)
    n_workers = max(1, n_workers)

    if (not USE_PARALLEL_MC) or n_workers == 1:
        rng = np.random.default_rng()
        for ii in range(ntest):
            if mc_method == 1:
                nfake = rng.poisson(float(n_events))
            else:
                nfake = n_events

            if nfake <= 1:
                maxpow_fake[ii] = 0.0
                pbest_fake[ii]  = np.nan
                continue

            tf = t0_mc + np.sort(rng.uniform(low=0.0, high=dt_window, size=nfake))
            powf = rayleigh_power(omega, tf)

            if verify:
                powf2 = np.array([
                    (np.cos(w*tf).sum()**2.0 + np.sin(w*tf).sum()**2.0) / float(tf.size)
                    for w in omega
                ], dtype=float)
                if np.max(np.abs(powf - powf2)) > 1e-10:
                    print("Warning: >1e-10 discrepancy between two Rayleigh implementations.")

            imax = np.argmax(powf)
            maxpow_fake[ii] = powf[imax]
            pbest_fake[ii]  = p[imax]

            if (ii % 1000) == 0 and ii > 0:
                print(f"  MC progress: {ii}/{ntest}")
    else:
        base_seed = int(np.random.SeedSequence().entropy)
        base = ntest // n_workers
        rem  = ntest % n_workers
        chunk_sizes = [base + (1 if i < rem else 0) for i in range(n_workers)]

        worker_args = []
        for i, ntrials in enumerate(chunk_sizes):
            if ntrials == 0:
                continue
            seed = base_seed + i
            worker_args.append((seed, ntrials, float(t0_mc), float(dt_window), int(n_events), omega, p, mc_method))

        chunk_sizes = [n for n in chunk_sizes if n > 0]
        n_workers = len(chunk_sizes)

        ctx = _get_mp_context()
        start_method = ctx.get_start_method()
        print(f"[MC] Running {ntest} trials using {n_workers} workers (start method: {start_method})...")

        with ProcessPoolExecutor(max_workers=n_workers, mp_context=ctx) as executor:
            offset = 0
            for (chunk_max, chunk_pbest), ntrials in zip(executor.map(_mc_chunk_worker, worker_args), chunk_sizes):
                maxpow_fake[offset:offset + ntrials] = chunk_max
                pbest_fake[offset:offset + ntrials]  = chunk_pbest
                offset += ntrials

        print("[MC] Parallel Monte Carlo complete.")

    fake_succ = np.count_nonzero(maxpow_fake > powmax)
    pvalue = (fake_succ + 1.0) / (ntest + 1.0)

    print(f"[MC] Fake datasets with Power > Real Power: {fake_succ} / {ntest}")
    print(f"[MC] P-value (Fake Power > Real Power): {pvalue:.6f}  ({pvalue:.3e})")

    return {
        "period_grid_min": p,
        "freq_grid_per_min": f,
        "power_data": pow_data,
        "best_period_min": pbest,
        "best_power": powmax,
        "mc_max_power": maxpow_fake,
        "mc_best_periods_min": pbest_fake,
        "p_value": pvalue,
        "lc_time_min": lc_centers,
        "lc_counts": lc_counts,
        "wavelet_periods_min": wavelet_periods,
        "wavelet_power": wavelet_power
    }

# ──────────────────────────────────────────────────────────────────────────────
# MAIN
# ──────────────────────────────────────────────────────────────────────────────
if __name__ == "__main__":
    _ensure_dir(events_out_dir)
    _ensure_dir(WAVES_OUTDIR)
    _ensure_dir(JADE_OUTDIR)

    for obs_id in ident:
        times_dt_utc, times_cxc = load_combined_times(obs_id, data_dir)
        idx = np.argsort(times_cxc)
        sorted_raw = times_cxc[idx]

        gaps, gap_idx = find_gaps(sorted_raw, threshold=gap_threshold)
        segments, seg_times_iso = split_into_segments(sorted_raw, gap_idx)
        segments, seg_times_iso, selected_label = pick_target_segment(segments, seg_times_iso, obs_id)

        if len(segments) == 0:
            print(f"No valid segments for {obs_id}")
            continue

        for si, seg in enumerate(segments, start=1):
            seg_label = selected_label if selected_label is not None else si
            start_iso, end_iso = seg_times_iso[si - 1]

            print(f"\n=== {obs_id} | Segment {seg_label} | {start_iso} ➜ {end_iso} ===")

            # Segment start UTC datetime for XMM plotting
            seg_t0_dt = Time(seg[0], format='cxcsec', scale='tt').utc.to_datetime()

            # Tag used for outputs
            tag = f"{obs_id}_seg{seg_label}_{start_iso[:19].replace(':','').replace('-','')}"

            # 0) JADE segment analysis + f_pe(t) for this window (used by Waves overlays)
            fpe_series = None
            try:
                fpe_series = jade_make_segment_analysis_and_fpe(
                    start_iso, end_iso, tag=tag,
                    emin_ev=JADE_ENERGY_MIN_EV,
                    emax_ev=JADE_ENERGY_MAX_EV,
                    save_png=True
                )
            except Exception as e:
                print(f"[JADE] Failed for {tag}: {e}")
                fpe_series = None

            # pass the computed series directly
            try:
                plot_waves_whistler(start_iso, end_iso, title_tag=tag,
                                    savepath=os.path.join(WAVES_OUTDIR, f"WAVES_WHIS_{tag}.png"),
                                    fpe_series=fpe_series)
            except Exception as e:
                print(f"[Waves] Failed for {tag}: {e}")


            # 2) Output unbinned XMM events
            rel_sec = save_unbinned_events(obs_id, seg_label, seg)

            # 3) XMM Rayleigh + MC + Wavelet + optional CML
            fig_tag = f"{obs_id} seg {seg_label} ({start_iso[:10]})"
            _ = run_rayleigh_and_mc(
                rel_sec,
                abs_t0=seg_t0_dt,
                startp_min=startp_min,
                endp_min=endp_min,
                nstep=nstep,
                time_convert=time_convert,
                time_restrict=time_restrict,
                tstart_min=tstart_min,
                tend_min=tend_min,
                ntest=ntest,
                mc_method=mc_method,
                verify=verify,
                plotit=plotit,
                fig_tag=fig_tag,
                lc_bin_min=lc_bin_min,
                n_scales_wavelet=80,
                plot_cml_overlay=plot_cml
            )

        print(f"Finished processing {obs_id}")

    print("All segments processed.")
```
