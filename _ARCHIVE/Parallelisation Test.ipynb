```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter
from astropy.table import Table
from astropy.time import Time
from astropy.coordinates import get_body_barycentric, solar_system_ephemeris
from astropy.constants import c
import astropy.units as u
import datetime as dt
import warnings
import os
import math

import pywt          # wavelets
import spiceypy as spice

import multiprocessing
from concurrent.futures import ProcessPoolExecutor

warnings.filterwarnings("ignore")

# ──────────────────────────────────────────────────────────────────────────────
# CONFIG
# ──────────────────────────────────────────────────────────────────────────────
ident = ['P0801930801', 'P0801930901', 'P0801931901', 'P0882110501', 'P0882110601']
TARGET_SEGMENTS = {
    'P0801930801': 2,
    'P0801930901': 1,
    'P0801931901': 3,
    'P0882110501': 3,
    'P0882110601': 5
}
'''
ident = ['P0801930801']
TARGET_SEGMENTS = { 'P0801930801': 2 }
'''
data_dir      = 'Data_Files/Fenn_Data/'
gap_threshold = 3600   # seconds

# Output locations
events_out_dir = r'/Users/fennleppard/Desktop/GitHub/X-Ray Pulsations/Segment Analysis Figures'

# ── Rayleigh periodogram / MC significance parameters (Knigge & Jackman style)
plotit         = True        # make diagnostic plots
verify         = False       # optional slow internal verification
ntest          = 2000       # Monte Carlo iterations
startp_min     = 1.0         # smallest period to search (minutes)
endp_min       = 15.0        # largest period to search (minutes)
nstep          = 1500        # number of frequency bins
time_convert   = True        # convert CXC seconds to minutes for analysis
time_restrict  = False       # restrict to [tstart, tend] in minutes (after conversion)
tstart_min     = 104.0       # start of window (min) if time_restrict=True
tend_min       = 454.0       # end of window (min) if time_restrict=True
mc_method      = 1           # 1=Poisson N per trial; 2=Fixed N per trial
lc_bin_min     = 0.25        # lightcurve bin size in minutes for wavelet

# Toggle for CML overlay
plot_cml       = True        # set False to disable CML overlay on lightcurve

# Color for CML curve
cml_color      = 'red'

# Multiprocessing defaults (tuned for Apple Silicon)
USE_PARALLEL_MC = True
PREFERRED_MP_START_METHOD = "fork"  # fastest on macOS; falls back to spawn if unavailable
TARGET_WORKERS = 10  # cap processes; adjust or set to None to use all available cores

# ──────────────────────────────────────────────────────────────────────────────
# SPICE HELPERS (CML)
# ──────────────────────────────────────────────────────────────────────────────
KERNEL_DIR = "SPICE KERNELS"
OBSERVER   = "EARTH"

def _furnish_dir(dirpath: str):
    if not dirpath or not os.path.isdir(dirpath):
        return
    for fn in sorted(os.listdir(dirpath)):
        if fn.lower().endswith((".tls", ".tpc", ".tf", ".tsc",
                                ".bsp", ".bc", ".cmt", ".tm")):
            try:
                spice.furnsh(os.path.join(dirpath, fn))
            except Exception:
                pass

# Load all kernels in the directory once
_furnish_dir(KERNEL_DIR)


def jupiter_cml_west_deg(et: float, observer: str = OBSERVER) -> float:
    """
    Compute Jupiter's Central Meridian Longitude (System III, West) in degrees
    as seen from the given observer at ephemeris time 'et'.
    """
    spoint, _, _ = spice.subpnt(
        method="Near Point: Ellipsoid",
        target="JUPITER",
        et=et,
        fixref="IAU_JUPITER",
        abcorr="LT+S",
        obsrvr=observer,
    )
    _, lon_east, _ = spice.reclat(spoint)
    cml_east = (np.degrees(lon_east) % 360.0)
    return float((360.0 - cml_east) % 360.0)


# ──────────────────────────────────────────────────────────────────────────────
# I/O & TIME CORRECTIONS
# ──────────────────────────────────────────────────────────────────────────────
def load_combined_times(obs_id, data_dir):
    """
    Read PN, M1, M2; correct to Jupiter–Earth light travel time; return:
      times_datetime (np.datetime64), times_cxcsec (float seconds, TT)
    """
    files = [f'{data_dir}EPIC_{obs_id}_{cam}_North_RFT.fit'
             for cam in ['PN', 'M1', 'M2']]
    list_dt, list_cxc = [], []

    for file in files:
        data = Table.read(file, format="fits")

        t = Time(data['TIME'], format='cxcsec', scale='tt')
        with solar_system_ephemeris.set('de432s'):
            jup = get_body_barycentric('jupiter', t)
            xmm = get_body_barycentric('earth', t)  # approx. XMM as Earth-based

        distance     = (jup - xmm).norm()                     # AU
        light_time_s = (distance.to('m') / c).to('s').value   # seconds
        corrected_t  = t - light_time_s * u.s

        list_dt.append(corrected_t.datetime)
        list_cxc.append(corrected_t.cxcsec)

    return np.hstack(list_dt), np.hstack(list_cxc)


def find_gaps(times_cxcsec, threshold=3600):
    dt_arr = np.diff(times_cxcsec)
    indices = np.where(dt_arr > threshold)[0]
    gaps    = [(times_cxcsec[i], times_cxcsec[i+1]) for i in indices]
    return gaps, indices


# ──────────────────────────────────────────────────────────────────────────────
# SEGMENT SELECTION & EVENT EXPORT (UNBINNED)
# ──────────────────────────────────────────────────────────────────────────────
def split_into_segments(sorted_times, gap_indices):
    split_pts = np.concatenate(([0], gap_indices + 1, [len(sorted_times)]))
    segments, seg_times_iso = [], []
    for i in range(len(split_pts) - 1):
        seg = sorted_times[split_pts[i]: split_pts[i+1]]
        if len(seg) > 1:
            segments.append(seg)
            seg_times_iso.append((
                Time(seg[0], format='cxcsec').iso,
                Time(seg[-1], format='cxcsec').iso
            ))
    return segments, seg_times_iso


def pick_target_segment(segments, seg_times_iso, obs_id):
    selected_1based = TARGET_SEGMENTS.get(obs_id)
    if selected_1based is None:
        return segments, seg_times_iso, None
    idx = selected_1based - 1
    if 0 <= idx < len(segments):
        return [segments[idx]], [seg_times_iso[idx]], selected_1based
    print(f"Warning: requested segment {selected_1based} for {obs_id} does not exist.")
    return [], [], selected_1based


def save_unbinned_events(obs_id, seg_idx_label, segment_times_cxc):
    """
    Save the raw (unbinned) corrected event times in CXC seconds.
    Also return a time series normalized to segment start for Rayleigh analysis.
    """
    out_txt = rf"{events_out_dir}/{obs_id}_seg_{seg_idx_label}_UNBINNED_EVENTS.txt"
    # Save as two columns: absolute CXC sec, relative (sec from segment start)
    t0 = segment_times_cxc[0]
    rel = segment_times_cxc - t0
    arr = np.column_stack([segment_times_cxc, rel])
    header = "ABS_CXC_SEC   REL_FROM_SEG_START_SEC"
    np.savetxt(out_txt, arr, fmt="%.6f", header=header)
    print(f"Saved unbinned events: {out_txt}")
    # Preview to stdout
    print("Preview (first 10 rows):")
    print(arr[:10])
    return rel  # seconds from segment start


# ──────────────────────────────────────────────────────────────────────────────
# RAYLEIGH PERIODOGRAM (KNIGGE & JACKMAN)
# ──────────────────────────────────────────────────────────────────────────────
use_numba = True
try:
    import numba
    from packaging.version import Version
    if Version(getattr(numba, "__version__", "0.0")) > Version("0.38"):
        use_numba = True
except Exception:
    use_numba = False


if use_numba:
    @numba.jit(nopython=True)
    def _rayleigh_power_numba(omega, tf):
        nfreq = omega.size
        powv = np.zeros(nfreq, dtype=np.float64)
        for i in range(nfreq):
            ssum = 0.0
            csum = 0.0
            w = omega[i]
            for k in range(tf.size):
                ang = w * tf[k]
                ssum += math.sin(ang)
                csum += math.cos(ang)
            powv[i] = (ssum*ssum + csum*csum) / float(tf.size)
        return powv


def rayleigh_power(omega, tf):
    """Return Rayleigh power spectrum for angular freqs omega and times tf."""
    if use_numba:
        return _rayleigh_power_numba(omega, tf)
    s2 = np.asarray([(np.sin(w*tf).sum())**2.0 for w in omega], dtype=float)
    c2 = np.asarray([(np.cos(w*tf).sum())**2.0 for w in omega], dtype=float)
    return (s2 + c2) / float(tf.size)


# Wavelet-based pulsation window helpers
def _find_high_power_intervals(wavelet_periods, wavelet_power, lc_time_min, target_period_min, min_frac=0.6):
    """Return [(t_start_min, t_end_min, peak_power), ...] where power near target period is high.
    min_frac sets the power threshold relative to the peak at that period.
    """
    if wavelet_power is None or wavelet_periods is None or lc_time_min is None:
        return []
    if len(wavelet_periods) == 0 or wavelet_power.size == 0 or lc_time_min.size == 0:
        return []
    idx = int(np.argmin(np.abs(wavelet_periods - target_period_min)))
    power_series = wavelet_power[idx]
    if power_series.size == 0:
        return []
    peak = float(np.max(power_series))
    if peak <= 0:
        return []
    thr = min_frac * peak
    intervals = []
    start_idx = None
    for i, val in enumerate(power_series):
        if val >= thr:
            if start_idx is None:
                start_idx = i
        elif start_idx is not None:
            seg_power = float(np.max(power_series[start_idx:i]))
            intervals.append((float(lc_time_min[start_idx]), float(lc_time_min[i-1]), seg_power))
            start_idx = None
    if start_idx is not None:
        seg_power = float(np.max(power_series[start_idx:]))
        intervals.append((float(lc_time_min[start_idx]), float(lc_time_min[len(power_series)-1]), seg_power))
    return intervals


def _plot_lightcurve_intervals(t_events_min, abs_t0, intervals, bin_sec=30.0):
    """Plot 30-second binned lightcurves for each high-power wavelet interval.
    t_events_min: event times (minutes from segment start)
    abs_t0: segment start datetime (or None)
    intervals: list of (t_start_min, t_end_min, peak_power)
    """
    if not intervals:
        print('[Wavelet] No high-power intervals found for per-interval lightcurve plots.')
        return
    bin_min = bin_sec / 60.0
    for j, (t0, t1, peak_pow) in enumerate(intervals, start=1):
        mask = (t_events_min >= t0) & (t_events_min <= t1)
        if not np.any(mask):
            continue
        tseg = t_events_min[mask]
        edges = np.arange(t0, t1 + bin_min, bin_min)
        if edges.size < 2:
            edges = np.array([t0, t0 + bin_min])
        counts, edges = np.histogram(tseg, bins=edges)
        centers = 0.5 * (edges[:-1] + edges[1:])

        if abs_t0 is not None:
            times_plot = np.array([abs_t0 + dt.timedelta(minutes=float(m)) for m in centers])
            xlabel = 'UTC Time'
        else:
            times_plot = centers
            xlabel = 'Time from seg start (min)'

        fig, ax = plt.subplots(figsize=(7, 3), dpi=140)
        ax.plot(times_plot, counts)
        ax.set_title(f'Lightcurve during pulsation window {j} (bin=30 s)')
        ax.set_xlabel(xlabel)
        ax.set_ylabel('Counts / 30 s')
        if abs_t0 is not None:
            ax.xaxis.set_major_formatter(DateFormatter('%H:%M:%S'))
            ax.tick_params(axis='x', rotation=45)
        ax.grid(True, alpha=0.3)
        plt.show()


# ──────────────────────────────────────────────────────────────────────────────
# MC WORKER FOR PARALLELISATION
# ──────────────────────────────────────────────────────────────────────────────
def _mc_chunk_worker(args):
    """
    Worker for a chunk of Monte Carlo trials.

    args = (
        seed,          # int
        ntrials,       # number of MC trials in this chunk
        t0_mc,         # float, window start (min)
        dt_window,     # float, window length (min)
        n_events,      # int, number of real events (size of t)
        omega,         # 1D array of angular frequencies
        p,             # 1D array of periods (same length as omega)
        mc_method,     # 1 or 2
    )
    """
    (seed, ntrials, t0_mc, dt_window, n_events, omega, p, mc_method) = args

    rng = np.random.default_rng(seed)

    maxpow_chunk = np.empty(ntrials, dtype=float)
    pbest_chunk  = np.empty(ntrials, dtype=float)

    for ii in range(ntrials):
        # Number of fake events
        if mc_method == 1:
            lam = float(n_events)
            nfake = rng.poisson(lam)
        else:
            nfake = n_events

        # Guard against nfake = 0 or 1 (can happen with Poisson)
        if nfake <= 1:
            maxpow_chunk[ii] = 0.0
            pbest_chunk[ii]  = np.nan
            continue

        # Uniform fake times in [t0_mc, t1_mc]
        tf = t0_mc + np.sort(rng.uniform(low=0.0, high=dt_window, size=nfake))

        # Rayleigh power for this fake dataset
        powf = rayleigh_power(omega, tf)

        imax = np.argmax(powf)
        maxpow_chunk[ii] = powf[imax]
        pbest_chunk[ii]  = p[imax]

    return maxpow_chunk, pbest_chunk


def _get_mp_context():
    """Choose a start method that works on macOS/Apple Silicon.
    Prefer 'fork' for speed; fall back to 'spawn' if unavailable.
    """
    available = set(multiprocessing.get_all_start_methods())
    method = PREFERRED_MP_START_METHOD if PREFERRED_MP_START_METHOD in available else "spawn"
    return multiprocessing.get_context(method)


# ──────────────────────────────────────────────────────────────────────────────
# MAIN ANALYSIS: RAYLEIGH + WAVELET + PLOTTING
# ──────────────────────────────────────────────────────────────────────────────
def run_rayleigh_and_mc(t_rel_seconds,
                        abs_t0=None,
                        startp_min=2.0, endp_min=100.0, nstep=1500,
                        time_convert=True, time_restrict=False,
                        tstart_min=0.0, tend_min=0.0,
                        ntest=10000, mc_method=2,
                        verify=False, plotit=True,
                        fig_tag="",
                        lc_bin_min=1.0,
                        n_scales_wavelet=80,
                        plot_cml_overlay=False):
    """
    t_rel_seconds: 1D array of event times relative to segment start (seconds)
    abs_t0:       Python datetime of segment start (UTC), for plotting in real time.
    All period params are in minutes.

    lc_bin_min:       lightcurve bin width in minutes
    n_scales_wavelet: number of periods/scales to sample between startp_min and endp_min
    plot_cml_overlay: if True, overlay Jupiter's CML on the lightcurve (right y-axis)
    """
    # Convert to minutes if desired
    t = np.array(t_rel_seconds, dtype=float)
    if time_convert:
        t = t / 60.0  # seconds -> minutes

    # Optional manual window restriction in minutes
    if time_restrict:
        mask = (t >= tstart_min) & (t <= tend_min)
        t = t[mask]
        if t.size < 2:
            raise ValueError("After restriction, fewer than 2 events remain.")

    # ──────────────────────────────────────────────────────────────
    # LIGHTCURVE (evenly sampled) for wavelet analysis
    # ──────────────────────────────────────────────────────────────
    if t.size > 0:
        t0_lc, t1_lc = t.min(), t.max()
        if t1_lc == t0_lc:
            t1_lc = t0_lc + lc_bin_min

        nbins = max(1, int(np.ceil((t1_lc - t0_lc) / lc_bin_min)))
        lc_edges   = np.linspace(t0_lc, t1_lc, nbins + 1)
        lc_counts, _ = np.histogram(t, bins=lc_edges)
        lc_centers = 0.5 * (lc_edges[:-1] + lc_edges[1:])
    else:
        lc_centers = np.array([])
        lc_counts  = np.array([])

    # Convert LC centers (minutes from segment start) → absolute UTC datetimes
    if abs_t0 is not None and lc_centers.size > 0:
        lc_datetimes = np.array(
            [abs_t0 + dt.timedelta(minutes=float(m)) for m in lc_centers]
        )
    else:
        lc_datetimes = None

    # ──────────────────────────────────────────────────────────────
    # RAYLEIGH PERIODOGRAM
    # ──────────────────────────────────────────────────────────────
    fmin = 1.0 / endp_min
    fmax = 1.0 / startp_min
    f = np.linspace(fmin, fmax, int(nstep))
    p = 1.0 / f
    omega = 2.0 * np.pi * f

    pow_data = rayleigh_power(omega, t)
    ibest   = np.argmax(pow_data)
    fbest   = f[ibest]
    pbest   = p[ibest]
    powmax  = pow_data[ibest]

    print(f"[Rayleigh] Best period = {pbest:.4f} min | best freq = {fbest:.6f} 1/min | Max power = {powmax:.6f}")

    # ──────────────────────────────────────────────────────────────
    # WAVELET TRANSFORM on lightcurve (where is that period in time?)
    # ──────────────────────────────────────────────────────────────
    wavelet_periods = None
    wavelet_power   = None

    if lc_centers.size > 3:
        wavename = 'morl'
        dt_min = lc_bin_min  # sampling period in minutes
        wavelet_periods = np.linspace(startp_min, endp_min, n_scales_wavelet)

        fc = pywt.central_frequency(wavename)
        scales = fc * wavelet_periods / dt_min

        y = lc_counts.astype(float) - np.mean(lc_counts)

        coeffs, freqs = pywt.cwt(y, scales, wavename, sampling_period=dt_min)
        wavelet_power = np.abs(coeffs)**2   # [n_scales, n_time]
    else:
        print("[Wavelet] Not enough lightcurve bins for wavelet analysis.")

    # ──────────────────────────────────────────────────────────────
    # PLOTTING
    # ──────────────────────────────────────────────────────────────
    if plotit:
        nrows = 4 if wavelet_power is not None else 3
        fig, axes = plt.subplots(nrows, 1, figsize=(8, 3*nrows + 2), dpi=150)

        ax_lc    = axes[0]
        ax_per   = axes[1]
        ax_freq  = axes[2]
        ax_wvlt  = axes[3] if nrows == 4 else None

        # 1) LIGHTCURVE (with optional CML overlay)
        if lc_counts.size > 0:
            if lc_datetimes is not None:
                ax_lc.step(lc_datetimes, lc_counts, where='mid')
                ax_lc.set_xlabel('UTC Time')
                ax_lc.xaxis.set_major_formatter(DateFormatter("%H:%M"))
                ax_lc.tick_params(axis='x', rotation=45)
            else:
                ax_lc.step(lc_centers, lc_counts, where='mid')
                ax_lc.set_xlabel('Time from seg start (min)')

            ax_lc.set_ylabel('Counts / bin')
            ax_lc.set_title(f'Lightcurve {fig_tag}')

            # ── CML overlay on right y-axis ─────────────────────────
            if plot_cml_overlay and (lc_datetimes is not None):
                try:
                    lc_et = np.array([spice.datetime2et(t) for t in lc_datetimes])
                    lc_cml = np.array([jupiter_cml_west_deg(et) for et in lc_et])

                    ax_cml = ax_lc.twinx()
                    ax_cml.plot(lc_datetimes, lc_cml,
                                color=cml_color, lw=1.6, alpha=0.5, linestyle='--')
                    ax_cml.set_ylabel("CML (°)", color=cml_color)
                    ax_cml.tick_params(axis='y', labelcolor=cml_color)
                    ax_cml.grid(False)
                    ax_cml.xaxis.set_major_formatter(DateFormatter("%H:%M"))
                    ax_cml.tick_params(axis='x', rotation=45)
                except Exception as e:
                    print(f"[CML] Failed to compute CML overlay: {e}")

        else:
            ax_lc.text(0.5, 0.5, 'No events', ha='center', va='center',
                       transform=ax_lc.transAxes)
            ax_lc.set_axis_off()

        # 2) RAYLEIGH POWER VS PERIOD
        ax_per.plot(p, pow_data, '-k', lw=1)
        ax_per.scatter(p, pow_data, s=12)
        ax_per.set_xlabel('Period (min)')
        ax_per.set_ylabel('Rayleigh Power')
        ax_per.set_title(f'Rayleigh Periodogram {fig_tag}')
        ax_per.axvline(pbest, ls='--', lw=1, color='gray')
        ax_per.annotate(f'{pbest:.2f} min',
                        xy=(pbest, powmax),
                        xytext=(pbest, powmax * 1.05),
                        ha='center', fontsize=8)
        ax_per.set_ylim(0, powmax * 1.2)

        # 3) RAYLEIGH POWER VS FREQUENCY
        ax_freq.plot(f, pow_data, '-k', lw=1)
        ax_freq.scatter(f, pow_data, s=12)
        ax_freq.set_xlabel('Frequency (1/min)')
        ax_freq.set_ylabel('Rayleigh Power')
        ax_freq.set_title(f'Rayleigh Periodogram (frequency) {fig_tag}')
        ax_freq.axvline(fbest, ls='--', lw=1, color='gray')
        ax_freq.annotate(f'{fbest:.4f} 1/min',
                         xy=(fbest, powmax),
                         xytext=(fbest, powmax * 1.05),
                         ha='center', fontsize=8)
        ax_freq.set_ylim(0, powmax * 1.2)

        # 4) WAVELET SCALOGRAM
        if ax_wvlt is not None and wavelet_power is not None:
            if lc_datetimes is not None:
                T, Pgrid = np.meshgrid(lc_datetimes, wavelet_periods)
            else:
                T, Pgrid = np.meshgrid(lc_centers, wavelet_periods)

            im = ax_wvlt.pcolormesh(T, Pgrid, wavelet_power,
                                    shading='auto', cmap='seismic')
            ax_wvlt.set_ylabel('Period (min)')
            if lc_datetimes is not None:
                ax_wvlt.set_xlabel('UTC Time')
                ax_wvlt.xaxis.set_major_formatter(DateFormatter("%H:%M"))
                ax_wvlt.tick_params(axis='x', rotation=45)
            else:
                ax_wvlt.set_xlabel('Time from seg start (min)')

            ax_wvlt.set_title(f'Wavelet Power (Morlet) {fig_tag}')
            ax_wvlt.axhline(pbest, color='white', ls='--', lw=1)
            ax_wvlt.invert_yaxis()
            #cbar = fig.colorbar(im, orientation='vertical')
            #cbar.set_label("Wavelet Power")

        fig.tight_layout()
        plt.show()

    # ──────────────────────────────────────────────────────────────
    # MONTE CARLO SIGNIFICANCE  (PARALLEL)
    # ──────────────────────────────────────────────────────────────
    if time_restrict:
        t0_mc, t1_mc = tstart_min, tend_min
    else:
        t0_mc, t1_mc = t.min(), t.max()
    dt_window = t1_mc - t0_mc
    if dt_window <= 0:
        raise ValueError("Non-positive time window for Monte Carlo.")

    n_events = t.size
    if n_events < 2:
        raise ValueError("Need at least 2 events for Monte Carlo significance.")

    maxpow_fake = np.empty(ntest, dtype=float)
    pbest_fake  = np.empty(ntest, dtype=float)

    # Parallelisation settings
    use_parallel_mc = USE_PARALLEL_MC
    n_cpu = os.cpu_count() or 1
    target_workers = TARGET_WORKERS or n_cpu
    n_workers = min(target_workers, n_cpu, ntest)
    if n_workers < 1:
        n_workers = 1

    if (not use_parallel_mc) or n_workers == 1:
        # Fallback: original serial loop
        rng = np.random.default_rng()
        for ii in range(ntest):
            if mc_method == 1:
                lam = float(n_events)
                nfake = rng.poisson(lam)
            else:
                nfake = n_events

            if nfake <= 1:
                maxpow_fake[ii] = 0.0
                pbest_fake[ii]  = np.nan
                continue

            tf = t0_mc + np.sort(rng.uniform(low=0.0, high=dt_window, size=nfake))
            powf = rayleigh_power(omega, tf)

            if verify:
                powf2 = np.array([
                    (np.cos(w*tf).sum()**2.0 + np.sin(w*tf).sum()**2.0) / float(tf.size)
                    for w in omega
                ], dtype=float)
                if np.max(np.abs(powf - powf2)) > 1e-10:
                    print("Warning: >1e-10 discrepancy between two Rayleigh implementations.")

            imax = np.argmax(powf)
            maxpow_fake[ii] = powf[imax]
            pbest_fake[ii]  = p[imax]

            if (ii % 1000) == 0 and ii > 0:
                print(f"  MC progress: {ii}/{ntest}")
    else:
        # Parallel path: split ntest into chunks across workers
        base_seed = int(np.random.SeedSequence().entropy)

        base = ntest // n_workers
        rem  = ntest % n_workers
        chunk_sizes = [base + (1 if i < rem else 0) for i in range(n_workers)]
        worker_args = []
        for i, ntrials in enumerate(chunk_sizes):
            if ntrials == 0:
                continue
            seed = base_seed + i
            worker_args.append(
                (seed, ntrials, float(t0_mc), float(dt_window),
                 int(n_events), omega, p, mc_method)
            )
        chunk_sizes = [n for n in chunk_sizes if n > 0]
        n_workers = len(chunk_sizes)

        ctx = _get_mp_context()
        start_method = ctx.get_start_method()
        if n_workers == 0:
            raise RuntimeError("No Monte Carlo work assigned to any worker.")
        print(f"[MC] Running {ntest} trials using {n_workers} worker processes (start method: {start_method})...")

        with ProcessPoolExecutor(max_workers=n_workers, mp_context=ctx) as executor:
            offset = 0
            for (chunk_max, chunk_pbest), ntrials in zip(
                executor.map(_mc_chunk_worker, worker_args),
                chunk_sizes
            ):
                maxpow_fake[offset:offset + ntrials] = chunk_max
                pbest_fake[offset:offset + ntrials]  = chunk_pbest
                offset += ntrials

        print("[MC] Parallel Monte Carlo complete.")

    # Improved p-value estimator to avoid exact zeros
    fake_succ = np.count_nonzero(maxpow_fake > powmax)
    pvalue = (fake_succ + 1.0) / (ntest + 1.0)

    print(f"[MC] Fake datasets with Power > Real Power: {fake_succ} / {ntest}")
    print(f"[MC] P-value (Fake Power > Real Power): {pvalue:.6f}  ({pvalue:.3e})")

    # Post-run Monte Carlo summary
    q_levels = [0.9, 0.99, 0.999, 0.9999]
    mc_quants = np.quantile(maxpow_fake, q_levels) if maxpow_fake.size > 0 else None
    if mc_quants is not None:
        print('[MC] Power quantiles:')
        for lvl, qv in zip(q_levels, mc_quants):
            print(f'  q{lvl:.4f}: {qv:.6f}')
        ratio = powmax / mc_quants[-1] if mc_quants[-1] > 0 else np.inf
        print(f'[MC] Real peak vs q0.9999: {powmax:.6f} / {mc_quants[-1]:.6f} = {ratio:.2f}x')

    return {
        "period_grid_min": p,
        "freq_grid_per_min": f,
        "power_data": pow_data,
        "best_period_min": pbest,
        "best_power": powmax,
        "mc_max_power": maxpow_fake,
        "mc_best_periods_min": pbest_fake,
        "p_value": pvalue,
        "lc_time_min": lc_centers,
        "lc_counts": lc_counts,
        "wavelet_periods_min": wavelet_periods,
        "wavelet_power": wavelet_power
    }


# ──────────────────────────────────────────────────────────────────────────────
# MAIN
# ──────────────────────────────────────────────────────────────────────────────
if __name__ == "__main__":
    for obs_id in ident:
        # Load and sort corrected times
        times_dt, times_cxc = load_combined_times(obs_id, data_dir)
        idx = np.argsort(times_cxc)
        sorted_raw = times_cxc[idx]

        # Find continuous segments, apply TARGET_SEGMENTS selection
        gaps, gap_idx = find_gaps(sorted_raw, threshold=gap_threshold)
        segments, seg_times_iso = split_into_segments(sorted_raw, gap_idx)
        segments, seg_times_iso, selected_label = pick_target_segment(
            segments, seg_times_iso, obs_id
        )

        if len(segments) == 0:
            print(f"No valid segments for {obs_id}")
            continue

        # Process each selected segment
        for si, seg in enumerate(segments, start=1):
            seg_label = selected_label if selected_label is not None else si
            start_iso, end_iso = seg_times_iso[si - 1]
            print(f"\n=== {obs_id} | Segment {seg_label} | {start_iso} ➜ {end_iso} ===")

            # Segment start UTC datetime (using corrected CXC time)
            seg_t0_dt = Time(seg[0], format='cxcsec', scale='tt').to_datetime()

            # 1) OUTPUT UNBINNED EVENTS
            rel_sec = save_unbinned_events(obs_id, seg_label, seg)

            # 2) RUN RAYLEIGH + MC + WAVELET + CML
            fig_tag = f"{obs_id} seg {seg_label} ({start_iso[:10]})"
            results = run_rayleigh_and_mc(
                rel_sec,
                abs_t0=seg_t0_dt,
                startp_min=startp_min,
                endp_min=endp_min,
                nstep=nstep,
                time_convert=time_convert,
                time_restrict=time_restrict,
                tstart_min=tstart_min,
                tend_min=tend_min,
                ntest=ntest,
                mc_method=mc_method,
                verify=verify,
                plotit=plotit,
                fig_tag=fig_tag,
                lc_bin_min=lc_bin_min,
                n_scales_wavelet=80,
                plot_cml_overlay=plot_cml
            )

        print(f"Finished processing {obs_id}")

    print("All segments processed.")
```
