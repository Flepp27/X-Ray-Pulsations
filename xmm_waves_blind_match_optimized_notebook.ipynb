Run the code cell below in Jupyter. It defines the functions, runs the analysis, saves CSV/PNG outputs, and leaves `xmm_candidates` and `waves_results` in memory.


```python
import os
import warnings
warnings.filterwarnings("ignore")

from typing import Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pywt

from joblib import Parallel, delayed
from scipy.signal import find_peaks
from statsmodels.stats.multitest import multipletests

from astropy.table import Table
from astropy.time import Time
from astropy.timeseries import LombScargle
from astropy.coordinates import get_body_barycentric, solar_system_ephemeris
from astropy.constants import c
import astropy.units as u

from das2 import get_source


# =============================================================================
# USER SETTINGS
# =============================================================================

# ---------------- XMM settings ----------------
# Only these observations and only these target segments will be analysed.
ident = ['P0801930801', 'P0801930901', 'P0801931901', 'P0882110501', 'P0882110601']

TARGET_SEGMENTS = {
    'P0801930801': 2,
    'P0801930901': 1,
    'P0801931901': 3,
    'P0882110501': 3,
    'P0882110601': 5,
}

data_dir = 'Data_Files/Fenn_Data/'

base_bin_size = 10
# Initial bin size used to classify segment count level.
# Dynamic rebinning then uses base_bin_size, 2*base_bin_size, or 4*base_bin_size.

gap_threshold_s = 3600.0

xmm_period_min_min = 2.0
xmm_period_max_min = 15.0
xmm_nfreq = 5000

# Require LS detections to be at least 95% Monte Carlo significant.
xmm_alpha_raw = 0.05
xmm_n_mc = 100
xmm_null_mode = "poisson"   # "poisson" or "shuffle"

min_segment_duration_min = 30.0
min_cycles_required = 3.0
max_xmm_candidates_per_segment = 3


# ---------------- WAVES settings ----------------
SERVER_URL_WAVES = "tag:das2.org,2012:site:/uiowa/juno/wav/survey/das2"

waves_dataset = "lhrl"      # "lhrl" or "lfrh"

# Speed/science trade-off:
# Scout mode suggestion: rows_per_bin=8, cwt_n_periods=60, waves_n_surrogates=30
# Final mode suggestion: rows_per_bin=4, cwt_n_periods=100-120, waves_n_surrogates=200-500
rows_per_bin = 8

waves_pad_minutes = 30.0

wavelet_name = "cmor1.5-1.0"
cwt_n_periods = 60
waves_period_min_min = 1.0
waves_period_max_min = 15.0

normalize_mode = "fractional"   # "fractional", "zscore", "mean", or "none"
spike_threshold = 1e4

period_tolerance_frac = 0.10
period_tolerance_min_floor = 0.5

waves_n_surrogates = 30
waves_alpha_fdr = 0.10

target_metric = "integrated"     # "integrated" or "max"
min_persistent_cycles = 2.0      # currently reported indirectly via persistence fraction

# Use [0.0] for strict simultaneity. Lag scanning greatly increases runtime.
lag_grid_min = [0.0]
# Example lag scan:
# lag_grid_min = np.arange(-60, 61, 10).tolist()

# Control-period comparison.
n_control_periods = 20
random_seed = 42


# ---------------- Parallel settings ----------------
# Apple M3 Max has 10 performance cores.
# For WAVES/PyWavelets/NumPy-heavy work, threading often avoids large array
# serialization overhead compared with loky.
N_JOBS = 10
JOBLIB_BACKEND = "threading"   # "threading" recommended here; "loky" is safer but often slower
BATCH_SIZE = "auto"

# Parallelisation strategy:
# - XMM Monte Carlo still parallelises over MC realisations.
# - WAVES now parallelises over frequency bins.
# - WAVES surrogates are serial inside each frequency-bin worker to avoid nested
#   joblib overhead and worker oversubscription.


# ---------------- Output ----------------
output_dir = "xmm_waves_blind_match_outputs"
os.makedirs(output_dir, exist_ok=True)


# =============================================================================
# XMM LOADING AND BINNING
# =============================================================================

def load_combined_times(obs_id: str, data_dir: str):
    """
    Read all three EPIC cameras, correct for Jupiter-Earth light travel time,
    and return:
      times_datetime : ndarray[datetime]
      times_cxcsec   : ndarray[float]
    """
    files = [f'{data_dir}EPIC_{obs_id}_{cam}_North_RFT.fit' for cam in ['PN', 'M1', 'M2']]
    list_dt, list_cxc = [], []

    for file in files:
        if not os.path.exists(file):
            print(f"[WARN] Missing file: {file}")
            continue

        data = Table.read(file, format="fits")

        t = Time(data['TIME'], format='cxcsec', scale='tt')
        with solar_system_ephemeris.set('de432s'):
            jup = get_body_barycentric('jupiter', t)
            xmm = get_body_barycentric('earth', t)  # approximate XMM as Earth-based

        distance_au = (jup - xmm).norm()
        light_time_s = (distance_au.to('m') / c).to('s').value

        corrected_t = t - light_time_s * u.s
        list_dt.append(corrected_t.datetime)
        list_cxc.append(corrected_t.cxcsec)

    if len(list_cxc) == 0:
        return np.array([]), np.array([])

    times_dt = np.hstack(list_dt)
    times_cxc = np.hstack(list_cxc)

    order = np.argsort(times_cxc)
    return times_dt[order], times_cxc[order]


def find_gaps(times_cxcsec: np.ndarray, threshold: float = 3600.0):
    dt = np.diff(times_cxcsec)
    indices = np.where(dt > threshold)[0]
    gaps = [(times_cxcsec[i], times_cxcsec[i + 1]) for i in indices]
    return gaps, indices


def split_into_segments(times_cxcsec: np.ndarray, threshold: float = 3600.0):
    """
    Split photon event times into continuous segments separated by large gaps.
    Segment numbering used elsewhere is 1-based.
    """
    if len(times_cxcsec) == 0:
        return []

    _, gap_indices = find_gaps(times_cxcsec, threshold=threshold)

    starts = np.r_[0, gap_indices + 1]
    ends = np.r_[gap_indices + 1, len(times_cxcsec)]

    segments = []
    for i, j in zip(starts, ends):
        seg = times_cxcsec[i:j]
        if len(seg) > 0:
            segments.append(seg)

    return segments


def bin_photon_times(photon_times: np.ndarray, bin_size: float):
    start = float(np.nanmin(photon_times))
    end = float(np.nanmax(photon_times))
    bins = np.arange(start, end + bin_size, bin_size)
    counts, edges = np.histogram(photon_times, bins=bins)
    centers = edges[:-1] + 0.5 * bin_size
    return centers, counts


def categorize_segment(counts: np.ndarray, low_thresh: float = 400, high_thresh: float = 800):
    total = float(np.sum(counts))
    if total < low_thresh:
        return 'low'
    elif total < high_thresh:
        return 'medium'
    return 'high'


def dynamic_bin_size(category: str, base_bin: int = 120):
    if category == 'low':
        return base_bin * 4
    elif category == 'medium':
        return base_bin * 2
    return base_bin


def prepare_segment_bins(raw_segment: np.ndarray, base_bin: int):
    """
    1) initial bin with base_bin to categorise
    2) choose seg_bin via dynamic_bin_size
    3) re-bin with seg_bin
    """
    _, counts0 = bin_photon_times(raw_segment, base_bin)
    category = categorize_segment(counts0)

    seg_bin = dynamic_bin_size(category, base_bin)
    hist_centers, hist_counts = bin_photon_times(raw_segment, seg_bin)
    return seg_bin, hist_centers, hist_counts, category


# =============================================================================
# XMM PERIOD SEARCH
# =============================================================================

def _one_xmm_null_max_power(t_s, counts, freqs_hz, null_mode, seed):
    """
    One XMM null Lomb-Scargle maximum-power realisation.
    """
    rng = np.random.default_rng(seed)
    mu = np.nanmean(counts)

    if null_mode == "poisson":
        y_null = rng.poisson(mu, size=len(counts))
    elif null_mode == "shuffle":
        y_null = rng.permutation(counts)
    else:
        raise ValueError("null_mode must be 'poisson' or 'shuffle'")

    if np.nanstd(y_null) <= 0:
        return 0.0

    ls_null = LombScargle(t_s, y_null - np.nanmean(y_null))
    p_null = ls_null.power(freqs_hz)
    return float(np.nanmax(p_null))


def _xmm_mc_threshold_and_pvals(
    t_s: np.ndarray,
    counts: np.ndarray,
    freqs_hz: np.ndarray,
    observed_powers: np.ndarray,
    n_mc: int = 1000,
    null_mode: str = "poisson",
    rng: Optional[np.random.Generator] = None,
    n_jobs: int = 1,
):
    """
    Monte Carlo null for the maximum LS power over the searched frequency band.
    Parallelised over Monte Carlo realisations when n_jobs > 1.
    """
    if rng is None:
        rng = np.random.default_rng()

    seeds = rng.integers(0, 2**32 - 1, size=n_mc, dtype=np.uint32)

    if n_jobs == 1:
        max_powers = [
            _one_xmm_null_max_power(
                t_s=t_s,
                counts=counts,
                freqs_hz=freqs_hz,
                null_mode=null_mode,
                seed=int(seed),
            )
            for seed in seeds
        ]
    else:
        max_powers = Parallel(
            n_jobs=n_jobs,
            backend=JOBLIB_BACKEND,
            batch_size=BATCH_SIZE,
            verbose=0,
        )(
            delayed(_one_xmm_null_max_power)(
                t_s=t_s,
                counts=counts,
                freqs_hz=freqs_hz,
                null_mode=null_mode,
                seed=int(seed),
            )
            for seed in seeds
        )

    max_powers = np.asarray(max_powers, dtype=float)

    pvals = np.array([
        (1.0 + np.sum(max_powers >= p)) / (n_mc + 1.0)
        for p in observed_powers
    ])

    return max_powers, pvals


def find_xmm_candidate_periods_for_segment(
    obs_id: str,
    segment_index: int,
    photon_times_cxc: np.ndarray,
    base_bin_size: int = 10,
    period_min_min: float = 2.0,
    period_max_min: float = 15.0,
    nfreq: int = 5000,
    alpha_raw: float = 0.05,
    n_mc: int = 1000,
    null_mode: str = "poisson",
    max_candidates: int = 3,
    min_duration_min: float = 30.0,
    min_cycles_required: float = 3.0,
    rng: Optional[np.random.Generator] = None,
):
    """
    Independently identify candidate XMM periods in one continuous segment.
    Candidate periods must pass the raw Monte Carlo alpha threshold.
    """
    if rng is None:
        rng = np.random.default_rng()

    duration_s = float(np.nanmax(photon_times_cxc) - np.nanmin(photon_times_cxc))
    duration_min = duration_s / 60.0

    if duration_min < min_duration_min:
        return pd.DataFrame()

    seg_bin, t_cxc, counts, category = prepare_segment_bins(photon_times_cxc, base_bin_size)

    if len(t_cxc) < 5 or np.nanstd(counts) <= 0:
        return pd.DataFrame()

    t_s = t_cxc - t_cxc[0]
    y = counts - np.nanmean(counts)

    f_min = 1.0 / (period_max_min * 60.0)
    f_max = 1.0 / (period_min_min * 60.0)
    freqs_hz = np.linspace(f_min, f_max, nfreq)

    ls = LombScargle(t_s, y)
    power = ls.power(freqs_hz)
    periods_min = 1.0 / freqs_hz / 60.0

    peaks, _ = find_peaks(power)
    if len(peaks) == 0:
        return pd.DataFrame()

    peak_powers = power[peaks]
    order = np.argsort(peak_powers)[::-1]
    peaks = peaks[order]
    peak_powers = peak_powers[order]

    # Optimisation: only Monte-Carlo test the strongest few peaks.
    # This avoids computing p-values for every local maximum when we only keep
    # max_candidates in the output.
    n_test_peaks = min(len(peaks), max_candidates * 5)
    peaks = peaks[:n_test_peaks]
    peak_powers = peak_powers[:n_test_peaks]

    _, peak_pvals_all = _xmm_mc_threshold_and_pvals(
        t_s=t_s,
        counts=counts,
        freqs_hz=freqs_hz,
        observed_powers=peak_powers,
        n_mc=n_mc,
        null_mode=null_mode,
        rng=rng,
        n_jobs=N_JOBS,
    )

    rows = []

    for rank, pk, pk_power, p_raw in zip(np.arange(1, len(peaks) + 1), peaks, peak_powers, peak_pvals_all):
        period_min = periods_min[pk]
        n_cycles = duration_min / period_min

        if p_raw > alpha_raw:
            continue

        if n_cycles < min_cycles_required:
            continue

        # Simple half-power width estimate in period space.
        half_power = 0.5 * pk_power

        left = pk
        while left > 0 and power[left] > half_power:
            left -= 1

        right = pk
        while right < len(power) - 1 and power[right] > half_power:
            right += 1

        p_left = periods_min[right] if right < len(periods_min) else period_min
        p_right = periods_min[left] if left >= 0 else period_min

        period_lo = min(p_left, p_right)
        period_hi = max(p_left, p_right)

        start_cxc = float(np.nanmin(photon_times_cxc))
        end_cxc = float(np.nanmax(photon_times_cxc))

        start_dt = Time(start_cxc, format='cxcsec', scale='tt').datetime
        end_dt = Time(end_cxc, format='cxcsec', scale='tt').datetime

        rows.append({
            "obs_id": obs_id,
            "segment_index": segment_index,
            "xmm_start_cxc": start_cxc,
            "xmm_end_cxc": end_cxc,
            "xmm_start_datetime": start_dt,
            "xmm_end_datetime": end_dt,
            "duration_min": duration_min,
            "total_counts": int(np.sum(counts)),
            "xmm_category": category,
            "xmm_bin_size_s": seg_bin,
            "xmm_period_min": float(period_min),
            "xmm_period_lo_min": float(period_lo),
            "xmm_period_hi_min": float(period_hi),
            "xmm_ls_power": float(pk_power),
            "xmm_p_raw": float(p_raw),
            "xmm_rank_in_segment": int(rank),
            "xmm_n_cycles": float(n_cycles),
        })

        if len(rows) >= max_candidates:
            break

    return pd.DataFrame(rows)


def find_all_xmm_candidates():
    """
    Analyse only the selected TARGET_SEGMENTS for the selected observations.
    """
    rng = np.random.default_rng(random_seed)
    all_rows = []

    for obs_id in ident:
        print(f"\n[XMM] Loading {obs_id}")
        _, times_cxc = load_combined_times(obs_id, data_dir)

        if len(times_cxc) == 0:
            print(f"[XMM] No valid events for {obs_id}")
            continue

        segments = split_into_segments(times_cxc, threshold=gap_threshold_s)
        print(f"[XMM] Found {len(segments)} segments")

        if obs_id not in TARGET_SEGMENTS:
            print(f"[XMM] No target segment specified for {obs_id}; skipping.")
            continue

        target_seg_index = TARGET_SEGMENTS[obs_id]

        if target_seg_index < 1 or target_seg_index > len(segments):
            print(
                f"[XMM] Target segment {target_seg_index} for {obs_id} "
                f"is out of range; valid range is 1-{len(segments)}. Skipping."
            )
            continue

        seg = segments[target_seg_index - 1]
        print(f"[XMM] Analysing only target segment {target_seg_index}")

        df_seg = find_xmm_candidate_periods_for_segment(
            obs_id=obs_id,
            segment_index=target_seg_index,
            photon_times_cxc=seg,
            base_bin_size=base_bin_size,
            period_min_min=xmm_period_min_min,
            period_max_min=xmm_period_max_min,
            nfreq=xmm_nfreq,
            alpha_raw=xmm_alpha_raw,
            n_mc=xmm_n_mc,
            null_mode=xmm_null_mode,
            max_candidates=max_xmm_candidates_per_segment,
            min_duration_min=min_segment_duration_min,
            min_cycles_required=min_cycles_required,
            rng=rng,
        )

        if len(df_seg) > 0:
            print(
                f"[XMM] {obs_id} seg {target_seg_index}: "
                f"{len(df_seg)} candidate(s): "
                f"{df_seg['xmm_period_min'].round(2).tolist()} min"
            )
            all_rows.append(df_seg)
        else:
            print(f"[XMM] {obs_id} seg {target_seg_index}: no >=95% LS candidates found")

    if len(all_rows) == 0:
        return pd.DataFrame()

    out = pd.concat(all_rows, ignore_index=True)

    # FDR correction across only the selected target-segment XMM candidate peaks.
    reject, p_corr, _, _ = multipletests(
        out["xmm_p_raw"].values,
        alpha=xmm_alpha_raw,
        method="fdr_bh",
    )

    out["xmm_p_fdr"] = p_corr
    out["xmm_detected_fdr"] = reject

    return out


# =============================================================================
# WAVES LOADING
# =============================================================================

def _ensure_dt64_ms(t):
    return np.array(t, dtype="datetime64[ms]")


def _parse_waves_dataset(d):
    """
    Extract times, freqs, spec from one DAS2 dataset block.
    Returns:
        times : datetime64[ms], shape (ntime,)
        freqs : float, shape (nfreq,)
        spec  : float, shape (ntime, nfreq)
    """
    f_arr = d["frequency"]["center"].array
    freqs = f_arr[0, :] if getattr(f_arr, "ndim", 1) > 1 else f_arr

    t_arr = d["Time"]["center"].array
    times = t_arr[:, 0] if getattr(t_arr, "ndim", 1) > 1 else t_arr

    spec = d["spec_dens"]["center"].array

    times = _ensure_dt64_ms(times)
    freqs = np.asarray(freqs, dtype=float)
    spec = np.asarray(spec, dtype=float)

    return times, freqs, spec


def _download_waves_segment(start_time, end_time, server_url=SERVER_URL_WAVES, dataset="lhrl"):
    """
    Download WAVES data for one interval.

    dataset:
        'lhrl' -> low-frequency/whistler dataset, ds[0]
        'lfrh' -> middle dataset, ds[1]
    """
    source = get_source(server_url)
    ds = source.get({"time": (start_time, end_time)})

    if len(ds) < 2:
        raise RuntimeError("Expected at least two datasets from DAS2.")

    if dataset.lower() == "lhrl":
        return _parse_waves_dataset(ds[0])
    elif dataset.lower() == "lfrh":
        return _parse_waves_dataset(ds[1])
    else:
        raise ValueError("dataset must be 'lhrl' or 'lfrh'")


def datetime_to_np64(dt):
    return np.datetime64(pd.Timestamp(dt).to_datetime64())


# =============================================================================
# WAVES PREPROCESSING AND CWT
# =============================================================================

def normalize_series(y, mode="fractional"):
    y = np.asarray(y, dtype=float)

    finite = np.isfinite(y)
    if np.sum(finite) < 3:
        return np.full_like(y, np.nan, dtype=float)

    mean = np.nanmean(y)
    std = np.nanstd(y)

    if mode == "fractional":
        if not np.isfinite(mean) or mean == 0:
            return np.full_like(y, np.nan, dtype=float)
        return (y - mean) / mean

    elif mode == "zscore":
        if not np.isfinite(std) or std == 0:
            return np.full_like(y, np.nan, dtype=float)
        return (y - mean) / std

    elif mode == "mean":
        if not np.isfinite(mean) or mean == 0:
            return np.full_like(y, np.nan, dtype=float)
        return y / mean

    elif mode == "none":
        return y.copy()

    else:
        raise ValueError("mode must be 'fractional', 'zscore', 'mean', or 'none'")


def interpolate_bad_values(y):
    y = np.asarray(y, dtype=float)
    x = np.arange(len(y))

    good = np.isfinite(y)
    if np.sum(good) < 3:
        return np.full_like(y, np.nan)

    return np.interp(x, x[good], y[good])


def remove_spikes(y, threshold=1e4):
    """
    Remove extremely large normalized spikes by interpolation.
    """
    y = np.asarray(y, dtype=float)
    bad = ~np.isfinite(y) | (np.abs(y) > threshold)

    if np.sum(~bad) < 3:
        return np.full_like(y, np.nan), int(np.sum(bad))

    y2 = y.copy()
    x = np.arange(len(y))
    y2[bad] = np.interp(x[bad], x[~bad], y[~bad])
    return y2, int(np.sum(bad))


def make_frequency_bins(freqs, rows_per_bin=4):
    bins = []
    nfreq = len(freqs)

    for i0 in range(0, nfreq, rows_per_bin):
        i1 = min(i0 + rows_per_bin, nfreq)
        bins.append((i0, i1, float(freqs[i0]), float(freqs[i1 - 1])))

    return bins


def summed_waves_series_for_freqbin(spec, i0, i1):
    """
    Sum WAVES spectral density over adjacent frequency rows.
    Input spec is expected shape (ntime, nfreq).
    """
    return np.nansum(spec[:, i0:i1], axis=1)


def compute_cwt_power(
    times64,
    y,
    wavelet_name="cmor1.5-1.0",
    period_min_min=1.0,
    period_max_min=15.0,
    n_periods=120,
):
    """
    Compute CWT power as function of period and time.

    Returns:
      periods_min, power
    """
    times64 = np.asarray(times64)
    y = np.asarray(y, dtype=float)

    if len(times64) < 5:
        return None, None

    t_sec = (times64 - times64[0]) / np.timedelta64(1, "s")
    dt = np.nanmedian(np.diff(t_sec))

    if not np.isfinite(dt) or dt <= 0:
        return None, None

    periods_min = np.geomspace(period_min_min, period_max_min, n_periods)
    periods_s = periods_min * 60.0

    wavelet = pywt.ContinuousWavelet(wavelet_name)
    cf = pywt.central_frequency(wavelet)
    scales = cf * periods_s / dt

    coeff, _ = pywt.cwt(y, scales, wavelet, sampling_period=dt)
    power = np.abs(coeff) ** 2

    return periods_min, power


# =============================================================================
# WAVES SURROGATES AND TEST STATISTICS
# =============================================================================

def phase_randomized_surrogate(y, rng):
    """
    Phase-randomized surrogate preserving the approximate Fourier amplitude spectrum.
    """
    y = np.asarray(y, dtype=float)
    y_mean = np.nanmean(y)
    y = y - y_mean

    n = len(y)
    fft = np.fft.rfft(y)
    amps = np.abs(fft)

    random_phases = rng.uniform(0, 2 * np.pi, size=len(fft))
    random_phases[0] = 0.0

    if n % 2 == 0:
        random_phases[-1] = 0.0

    fft_new = amps * np.exp(1j * random_phases)
    y_surr = np.fft.irfft(fft_new, n=n)

    if np.nanstd(y_surr) > 0 and np.nanstd(y) > 0:
        y_surr = y_surr / np.nanstd(y_surr) * np.nanstd(y)

    y_surr = y_surr + y_mean
    return y_surr


def target_period_band(periods_min, target_period_min, tolerance_frac=0.10, floor_min=0.5):
    tol = max(target_period_min * tolerance_frac, floor_min)
    lo = target_period_min - tol
    hi = target_period_min + tol
    mask = (periods_min >= lo) & (periods_min <= hi)
    return mask, lo, hi


def waves_target_statistic(
    periods_min,
    power,
    times64,
    central_start64,
    central_end64,
    target_period_min,
    tolerance_frac=0.10,
    floor_min=0.5,
    metric="integrated",
):
    """
    Statistic comparing target-band CWT power with off-target CWT power.

    Returns:
      stat, target_power, background_power, persistence_fraction
    """
    if periods_min is None or power is None:
        return np.nan, np.nan, np.nan, np.nan

    periods_min = np.asarray(periods_min)
    times64 = np.asarray(times64)

    central_mask = (times64 >= central_start64) & (times64 <= central_end64)
    if np.sum(central_mask) < 5:
        return np.nan, np.nan, np.nan, np.nan

    band_mask, _, _ = target_period_band(
        periods_min,
        target_period_min,
        tolerance_frac=tolerance_frac,
        floor_min=floor_min,
    )

    if np.sum(band_mask) == 0:
        return np.nan, np.nan, np.nan, np.nan

    bg_mask = ~band_mask

    target_power_map = power[np.ix_(band_mask, central_mask)]
    bg_power_map = power[np.ix_(bg_mask, central_mask)]

    if target_power_map.size == 0 or bg_power_map.size == 0:
        return np.nan, np.nan, np.nan, np.nan

    if metric == "max":
        target_power = np.nanmax(target_power_map)
    elif metric == "integrated":
        target_power = np.nanmean(target_power_map)
    else:
        raise ValueError("metric must be 'max' or 'integrated'")

    background_power = np.nanmedian(bg_power_map)

    if not np.isfinite(background_power) or background_power <= 0:
        stat = np.nan
    else:
        stat = target_power / background_power

    bg_thr = np.nanpercentile(bg_power_map, 90)
    target_vs_time = np.nanmean(target_power_map, axis=0)
    persistence_fraction = np.nanmean(target_vs_time > bg_thr)

    return float(stat), float(target_power), float(background_power), float(persistence_fraction)


def _one_waves_surrogate_stat(
    y,
    times64,
    central_start64,
    central_end64,
    target_period_min,
    seed,
):
    """
    One phase-randomized surrogate WAVES statistic.
    This is kept serial inside each frequency-bin worker in the optimized version.
    """
    rng = np.random.default_rng(seed)
    y_surr = phase_randomized_surrogate(y, rng)

    periods_surr, power_surr = compute_cwt_power(
        times64,
        y_surr,
        wavelet_name=wavelet_name,
        period_min_min=waves_period_min_min,
        period_max_min=waves_period_max_min,
        n_periods=cwt_n_periods,
    )

    stat_surr, _, _, _ = waves_target_statistic(
        periods_surr,
        power_surr,
        times64,
        central_start64,
        central_end64,
        target_period_min,
        tolerance_frac=period_tolerance_frac,
        floor_min=period_tolerance_min_floor,
        metric=target_metric,
    )

    return stat_surr


def waves_surrogate_pvalue(
    times64,
    y,
    central_start64,
    central_end64,
    target_period_min,
    observed_stat,
    n_surrogates=500,
    rng=None,
    n_jobs=1,
):
    """
    Compute p-value for WAVES target-period statistic using phase-randomized
    surrogates.

    In this optimized script, use n_jobs=1 for this function when it is called
    inside the frequency-bin worker. The outer loop over frequency bins is the
    preferred parallelisation layer.
    """
    if rng is None:
        rng = np.random.default_rng()

    if not np.isfinite(observed_stat):
        return np.nan, np.array([])

    seeds = rng.integers(0, 2**32 - 1, size=n_surrogates, dtype=np.uint32)

    if n_jobs == 1:
        null_stats = [
            _one_waves_surrogate_stat(
                y=y,
                times64=times64,
                central_start64=central_start64,
                central_end64=central_end64,
                target_period_min=target_period_min,
                seed=int(seed),
            )
            for seed in seeds
        ]
    else:
        null_stats = Parallel(
            n_jobs=n_jobs,
            backend=JOBLIB_BACKEND,
            batch_size=BATCH_SIZE,
            verbose=0,
        )(
            delayed(_one_waves_surrogate_stat)(
                y=y,
                times64=times64,
                central_start64=central_start64,
                central_end64=central_end64,
                target_period_min=target_period_min,
                seed=int(seed),
            )
            for seed in seeds
        )

    null_stats = np.asarray(null_stats, dtype=float)
    null_stats = null_stats[np.isfinite(null_stats)]

    if len(null_stats) == 0:
        return np.nan, null_stats

    p = (1.0 + np.sum(null_stats >= observed_stat)) / (len(null_stats) + 1.0)
    return float(p), null_stats


def generate_control_periods(
    target_period_min,
    n_controls,
    rng,
    period_min=2.0,
    period_max=15.0,
    exclusion_frac=0.15,
):
    """
    Generate random control periods excluding a region around the XMM target period.
    """
    controls = []
    exclusion_width = target_period_min * exclusion_frac

    attempts = 0
    while len(controls) < n_controls and attempts < n_controls * 100:
        p = rng.uniform(period_min, period_max)
        if abs(p - target_period_min) > exclusion_width:
            controls.append(p)
        attempts += 1

    return controls


# =============================================================================
# MAIN XMM-WAVES BLIND TEST
# =============================================================================

def _test_one_waves_freq_bin(
    bin_index,
    i0,
    i1,
    f_lo,
    f_hi,
    waves_times,
    waves_spec,
    central_start64,
    central_end64,
    candidate_row_dict,
    control_periods,
    seed,
):
    """
    Test one WAVES frequency bin against one XMM candidate.

    Optimisation:
    - This function is called in parallel over frequency bins.
    - Surrogates are computed serially inside this worker to avoid nested
      joblib calls and repeated worker overhead.
    """
    rng = np.random.default_rng(seed)

    obs_id = candidate_row_dict["obs_id"]
    seg_idx = int(candidate_row_dict["segment_index"])
    target_period = float(candidate_row_dict["xmm_period_min"])

    y_raw = summed_waves_series_for_freqbin(waves_spec, i0, i1)
    y_norm = normalize_series(y_raw, normalize_mode)
    y_norm = interpolate_bad_values(y_norm)
    y_clean, n_spikes = remove_spikes(y_norm, threshold=spike_threshold)

    if np.sum(np.isfinite(y_clean)) < 10 or np.nanstd(y_clean) <= 0:
        return []

    periods_min, power = compute_cwt_power(
        waves_times,
        y_clean,
        wavelet_name=wavelet_name,
        period_min_min=waves_period_min_min,
        period_max_min=waves_period_max_min,
        n_periods=cwt_n_periods,
    )

    if periods_min is None:
        return []

    rows = []

    for lag_min in lag_grid_min:
        lag64 = np.timedelta64(int(lag_min * 60), "s")
        test_start64 = central_start64 + lag64
        test_end64 = central_end64 + lag64

        stat, target_power, bg_power, persistence = waves_target_statistic(
            periods_min,
            power,
            waves_times,
            test_start64,
            test_end64,
            target_period,
            tolerance_frac=period_tolerance_frac,
            floor_min=period_tolerance_min_floor,
            metric=target_metric,
        )

        p_raw, null_stats = waves_surrogate_pvalue(
            times64=waves_times,
            y=y_clean,
            central_start64=test_start64,
            central_end64=test_end64,
            target_period_min=target_period,
            observed_stat=stat,
            n_surrogates=waves_n_surrogates,
            rng=rng,
            n_jobs=1,
        )

        control_stats = []
        for cp in control_periods:
            cp_stat, _, _, _ = waves_target_statistic(
                periods_min,
                power,
                waves_times,
                test_start64,
                test_end64,
                cp,
                tolerance_frac=period_tolerance_frac,
                floor_min=period_tolerance_min_floor,
                metric=target_metric,
            )
            if np.isfinite(cp_stat):
                control_stats.append(cp_stat)

        control_stats = np.asarray(control_stats)

        if len(control_stats) > 0 and np.isfinite(stat):
            control_percentile = 100.0 * np.mean(control_stats <= stat)
            control_median = float(np.nanmedian(control_stats))
        else:
            control_percentile = np.nan
            control_median = np.nan

        _, p_lo, p_hi = target_period_band(
            periods_min,
            target_period,
            tolerance_frac=period_tolerance_frac,
            floor_min=period_tolerance_min_floor,
        )

        rows.append({
            "obs_id": obs_id,
            "segment_index": seg_idx,
            "xmm_start_datetime": candidate_row_dict["xmm_start_datetime"],
            "xmm_end_datetime": candidate_row_dict["xmm_end_datetime"],
            "xmm_period_min": target_period,
            "xmm_period_lo_min": candidate_row_dict["xmm_period_lo_min"],
            "xmm_period_hi_min": candidate_row_dict["xmm_period_hi_min"],
            "xmm_ls_power": candidate_row_dict["xmm_ls_power"],
            "xmm_p_raw": candidate_row_dict["xmm_p_raw"],
            "xmm_p_fdr": candidate_row_dict.get("xmm_p_fdr", np.nan),
            "waves_dataset": waves_dataset,
            "freq_bin_index": bin_index,
            "freq_i0": i0,
            "freq_i1": i1,
            "freq_lo_hz": f_lo,
            "freq_hi_hz": f_hi,
            "lag_min": lag_min,
            "target_band_lo_min": p_lo,
            "target_band_hi_min": p_hi,
            "waves_stat": stat,
            "waves_target_power": target_power,
            "waves_background_power": bg_power,
            "waves_persistence_fraction": persistence,
            "waves_p_raw": p_raw,
            "n_spikes_removed": n_spikes,
            "n_surrogates_used": len(null_stats),
            "control_median_stat": control_median,
            "control_percentile": control_percentile,
            "n_control_periods": len(control_stats),
        })

    return rows


def test_one_xmm_candidate_against_waves(candidate_row, rng):
    """
    For one XMM candidate period and interval:
      - Download padded WAVES data once.
      - Parallelise over WAVES frequency bins.
      - For each frequency bin and lag:
          - Compute observed target-period statistic.
          - Compute surrogate p-value.
      - Compute control-period statistics for comparison.
    """
    obs_id = candidate_row["obs_id"]
    seg_idx = int(candidate_row["segment_index"])
    target_period = float(candidate_row["xmm_period_min"])

    central_start_dt = candidate_row["xmm_start_datetime"]
    central_end_dt = candidate_row["xmm_end_datetime"]

    central_start64 = datetime_to_np64(central_start_dt)
    central_end64 = datetime_to_np64(central_end_dt)

    pad = np.timedelta64(int(waves_pad_minutes * 60), "s")
    download_start64 = central_start64 - pad
    download_end64 = central_end64 + pad

    download_start_str = str(download_start64).replace(".000000", "")
    download_end_str = str(download_end64).replace(".000000", "")

    print(
        f"[WAVES] {obs_id} seg {seg_idx}: "
        f"testing XMM period {target_period:.2f} min, "
        f"{download_start_str} to {download_end_str}"
    )

    try:
        waves_times, waves_freqs, waves_spec = _download_waves_segment(
            download_start_str,
            download_end_str,
            dataset=waves_dataset,
        )
    except Exception as e:
        print(f"[WAVES] Download failed for {obs_id} seg {seg_idx}: {e}")
        return pd.DataFrame()

    freq_bins = make_frequency_bins(waves_freqs, rows_per_bin=rows_per_bin)
    print(
        f"[WAVES] {obs_id} seg {seg_idx}: "
        f"{len(freq_bins)} frequency bins, {waves_n_surrogates} surrogates/bin, "
        f"CWT periods={cwt_n_periods}, backend={JOBLIB_BACKEND}"
    )

    control_periods = generate_control_periods(
        target_period,
        n_control_periods,
        rng,
        period_min=xmm_period_min_min,
        period_max=xmm_period_max_min,
    )

    candidate_row_dict = candidate_row.to_dict()

    seeds = rng.integers(
        0,
        2**32 - 1,
        size=len(freq_bins),
        dtype=np.uint32,
    )

    results_nested = Parallel(
        n_jobs=N_JOBS,
        backend=JOBLIB_BACKEND,
        batch_size=BATCH_SIZE,
        verbose=5,
    )(
        delayed(_test_one_waves_freq_bin)(
            bin_index=bin_index,
            i0=i0,
            i1=i1,
            f_lo=f_lo,
            f_hi=f_hi,
            waves_times=waves_times,
            waves_spec=waves_spec,
            central_start64=central_start64,
            central_end64=central_end64,
            candidate_row_dict=candidate_row_dict,
            control_periods=control_periods,
            seed=int(seed),
        )
        for seed, (bin_index, (i0, i1, f_lo, f_hi)) in zip(
            seeds,
            enumerate(freq_bins, start=1),
        )
    )

    rows = [row for sublist in results_nested for row in sublist]

    if len(rows) == 0:
        return pd.DataFrame()

    return pd.DataFrame(rows)


def run_blind_xmm_waves_matching():
    rng = np.random.default_rng(random_seed)

    print("\n==============================")
    print("STEP 1: Finding XMM candidates")
    print("==============================")

    xmm_candidates = find_all_xmm_candidates()

    if len(xmm_candidates) == 0:
        print("[DONE] No >=95% XMM candidate periods found in the selected target segments.")
        return pd.DataFrame(), pd.DataFrame()

    xmm_path = os.path.join(output_dir, "xmm_candidate_periods.csv")
    xmm_candidates.to_csv(xmm_path, index=False)
    print(f"[SAVE] XMM candidates: {xmm_path}")

    # Strict version: only test XMM candidates surviving 95% FDR correction.
    test_candidates = xmm_candidates[xmm_candidates["xmm_detected_fdr"]].copy()

    if len(test_candidates) == 0:
        print("[DONE] No XMM candidates survived 95% FDR correction. No WAVES tests run.")
        return xmm_candidates, pd.DataFrame()

    print("\n================================")
    print("STEP 2: Testing WAVES blindly")
    print("================================")

    all_waves_results = []

    for _, row in test_candidates.iterrows():
        df_waves = test_one_xmm_candidate_against_waves(row, rng)
        if len(df_waves) > 0:
            all_waves_results.append(df_waves)

    if len(all_waves_results) == 0:
        print("[DONE] No WAVES results produced.")
        return xmm_candidates, pd.DataFrame()

    waves_results = pd.concat(all_waves_results, ignore_index=True)

    valid = np.isfinite(waves_results["waves_p_raw"].values)
    waves_results["waves_p_fdr"] = np.nan
    waves_results["waves_detected_fdr"] = False

    if np.sum(valid) > 0:
        reject, p_corr, _, _ = multipletests(
            waves_results.loc[valid, "waves_p_raw"].values,
            alpha=waves_alpha_fdr,
            method="fdr_bh",
        )
        waves_results.loc[valid, "waves_p_fdr"] = p_corr
        waves_results.loc[valid, "waves_detected_fdr"] = reject

    waves_path = os.path.join(output_dir, "waves_blind_test_results.csv")
    waves_results.to_csv(waves_path, index=False)
    print(f"[SAVE] WAVES blind-test results: {waves_path}")

    return xmm_candidates, waves_results


# =============================================================================
# PLOTTING / SUMMARY FIGURES
# =============================================================================

def plot_waves_detection_summary(waves_results: pd.DataFrame):
    """
    Basic summary plot:
      - frequency-bin midpoint versus XMM period
      - significant points highlighted after WAVES FDR correction.
    """
    if len(waves_results) == 0:
        print("[PLOT] No WAVES results to plot.")
        return

    df = waves_results.copy()
    df["freq_mid_hz"] = np.sqrt(df["freq_lo_hz"] * df["freq_hi_hz"])
    detected = df["waves_detected_fdr"].fillna(False)

    plt.figure(figsize=(10, 6))

    plt.scatter(
        df.loc[~detected, "xmm_period_min"],
        df.loc[~detected, "freq_mid_hz"],
        s=25,
        alpha=0.35,
        label="Not WAVES FDR significant",
    )

    if np.sum(detected) > 0:
        plt.scatter(
            df.loc[detected, "xmm_period_min"],
            df.loc[detected, "freq_mid_hz"],
            s=70,
            alpha=0.9,
            label="WAVES FDR significant",
        )

    plt.yscale("log")
    plt.xlabel("XMM-derived period tested in WAVES [min]")
    plt.ylabel("WAVES frequency-bin midpoint [Hz]")
    plt.title("Blind XMM-period tests in WAVES CWT")
    plt.legend()
    plt.tight_layout()

    path = os.path.join(output_dir, "waves_detection_summary.png")
    plt.savefig(path, dpi=200)
    plt.show()

    print(f"[SAVE] {path}")


def plot_control_comparison(waves_results: pd.DataFrame):
    """
    Plot how the XMM-period WAVES statistic compares with random control periods.
    """
    if len(waves_results) == 0 or "control_percentile" not in waves_results.columns:
        print("[PLOT] No control-period results to plot.")
        return

    df = waves_results[np.isfinite(waves_results["control_percentile"])].copy()

    if len(df) == 0:
        print("[PLOT] No finite control percentiles.")
        return

    plt.figure(figsize=(8, 5))
    plt.hist(df["control_percentile"], bins=np.linspace(0, 100, 21), alpha=0.8)
    plt.axvline(95, linestyle="--", label="95th percentile")
    plt.xlabel("Percentile of XMM-period WAVES statistic relative to control periods")
    plt.ylabel("Number of WAVES tests")
    plt.title("Are XMM periods special relative to random control periods?")
    plt.legend()
    plt.tight_layout()

    path = os.path.join(output_dir, "control_period_percentiles.png")
    plt.savefig(path, dpi=200)
    plt.show()

    print(f"[SAVE] {path}")


def summarize_results(xmm_candidates: pd.DataFrame, waves_results: pd.DataFrame):
    print("\n================")
    print("SUMMARY")
    print("================")

    print(f"Total XMM candidate periods: {len(xmm_candidates)}")

    if len(xmm_candidates) > 0:
        print(f"XMM FDR-significant candidates at 95%: {int(xmm_candidates['xmm_detected_fdr'].sum())}")

    print(f"Total WAVES tests: {len(waves_results)}")

    if len(waves_results) > 0:
        print(f"WAVES FDR-significant tests: {int(waves_results['waves_detected_fdr'].sum())}")

        sig = waves_results[waves_results["waves_detected_fdr"]].copy()

        if len(sig) > 0:
            cols = [
                "obs_id", "segment_index", "xmm_period_min",
                "freq_lo_hz", "freq_hi_hz", "lag_min",
                "waves_stat", "waves_p_raw", "waves_p_fdr",
                "waves_persistence_fraction", "control_percentile",
            ]
            print("\nTop significant WAVES matches:")
            print(sig[cols].sort_values("waves_p_fdr").head(20).to_string(index=False))


# =============================================================================
# NOTEBOOK RUN CELL
# =============================================================================
# Run this final block in Jupyter after the definitions above have executed.
# The recommended backend is threading, so no multiprocessing start-method block
# is needed.

print(f"[INFO] Running with N_JOBS={N_JOBS}, backend={JOBLIB_BACKEND}")
print(f"[INFO] XMM MC realisations per segment: {xmm_n_mc}")
print(f"[INFO] WAVES surrogates per test: {waves_n_surrogates}")
print(f"[INFO] rows_per_bin={rows_per_bin}, cwt_n_periods={cwt_n_periods}")

xmm_candidates, waves_results = run_blind_xmm_waves_matching()

summarize_results(xmm_candidates, waves_results)

if len(waves_results) > 0:
    plot_waves_detection_summary(waves_results)
    plot_control_comparison(waves_results)

# In Jupyter, these variables remain available after the cell finishes:
#   xmm_candidates
#   waves_results
```
